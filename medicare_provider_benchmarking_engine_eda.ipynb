### **Data - RBCS Taxonomy**

The Restructured Berenson-Eggers Type of Service (RBCS) is a CMS-maintained, updated taxonomy (successor to BETOS 1.0) that categorizes Medicare Part B HCPCS codes into clinically meaningful groups to analyze spending, utilization, and trends. It provides a hierarchical structure (categories, subcategories, families) designed to be stable over time, covering both physician and non-physician services. 

Key Aspects of the RBCS:
- Purpose: Developed for the Centers for Medicare & Medicaid Services (CMS) to analyze Medicare Part B expenditures and trends.
- Structure: Hierarchically organized into categories (highest level), subcategories, and families, ensuring consistent, clinically meaningful groupings.
- Mapping: RBCS IDs are 6-character codes identifying the category, subcategory, family, and whether the service is a major procedure (M), other (O), or non-procedure (N).
- Updates: The system is updated annually by a technical expert panel to reflect modern clinical practice, replacing the outdated 1980s-era original BETOS.
- Coverage: It maps all Medicare Part B services, including those from the physician fee schedule, expanding on the scope of previous versions. 

The RBCS is crucial for policy, research, and tracking how changes in coding or technology affect Medicare costs

***CRITICAL NOTE: At the end of "EDA 6. Specialty comparisons (our 5 oncology types)" I discovered that the original `rbcs` dataset has duplicates that led to fatal downstream duplication.*** 

The code below fixes the `rbcs` dataset by removing duplicates and generating a clean dataset called `rbcs_dedup`. 

We will rebuild the `prov_svc_onco_core_prov_all_rbcs` dataset with the clean `rbcs_dedup` and run the entire notebook from start to finish to ensure the duplication issue is completely resolved.

````python
con.execute(
    """
CREATE OR REPLACE VIEW rbcs_dedup AS
SELECT *
FROM (
  SELECT
    r.*,
    ROW_NUMBER() OVER (
      PARTITION BY HCPCS_Cd
      ORDER BY
        RBCS_Latest_Assignment DESC,
        RBCS_Analysis_Start_Dt DESC NULLS LAST,
        RBCS_Analysis_End_Dt DESC NULLS LAST,
        CASE WHEN RBCS_FamNumb = '000' THEN 1 ELSE 0 END
    ) AS rn
  FROM rbcs r
)
WHERE rn = 1;
"""
)
````

After deduping, confirm RBCS is now unique on HCPCS:

````python
con.sql(
    """
SELECT
  COUNT(*) AS rbcs_dedup_rows,
  COUNT(DISTINCT HCPCS_Cd) AS distinct_hcpcs
FROM rbcs_dedup;
"""
)
````

### **Data - NPPES**

The National Plan and Provider Enumeration System (NPPES) is a CMS-maintained database that assigns unique 10-digit National Provider Identifiers (NPIs) to health care providers. Essential for Medicare and Medicaid, this system ensures standardized, digital identification for billing, claims processing, and compliance with HIPAA regulations. 

Key Aspects of NPPES for Medicare/Medicaid:

- Purpose: NPPES stores provider data—name, address, taxonomy, and taxonomy—to uniquely identify providers.
- Mandate: Under HIPAA, all providers must have an NPI to submit claims to Medicare, Medicaid, and private payers.
- Application Process: Providers must create an Identity & Access (I&A) account to apply for an NPI, which is the fastest method, or mail a paper form.
- Maintenance: Providers must keep their NPPES information up to date, as incorrect information can cause claims to be rejected.
- NPI vs. Medicare Number: An NPI is not the same as a Medicare Provider Transaction Access Number (PTAN); the NPI is national, while the Medicare number is specific to the contractor.

#### Let's left join the `prov_svc_onco_core_prov_all_rbcs` table with the `nppes` table on `Rndrng_NPI` on the left side and `NPI` on the right side.

````python
con.execute(
    """
CREATE OR REPLACE VIEW prov_svc_onco_core_prov_all_rbcs_nppes AS
SELECT
    t.*,
    n.enumeration_date,
    CASE
        WHEN n.enumeration_date IS NULL
        THEN NULL
        ELSE (t.Year - EXTRACT(year FROM n.enumeration_date))::INTEGER
    END AS years_since_enumeration
FROM prov_svc_onco_core_prov_all_rbcs AS t
LEFT JOIN nppes AS n
    ON CAST(t.Rndrng_NPI AS BIGINT) = n.NPI;
"""
)
````

````python
con.sql(
    """
SELECT
  COUNT(*) AS n_rows,
  AVG(CASE WHEN enumeration_date IS NULL THEN 1 ELSE 0 END) AS pct_missing_enumeration_date,
  MIN(years_since_enumeration) AS min_years,
  MAX(years_since_enumeration) AS max_years
FROM prov_svc_onco_core_prov_all_rbcs_nppes;
"""
)
````

````python
con.execute(
    """
CREATE OR REPLACE TABLE prov_svc_onco_core_prov_all_rbcs_nppes_tbl AS
SELECT * FROM prov_svc_onco_core_prov_all_rbcs_nppes;
"""
)
````

````python
OUT_PARQUET = (DATA / "prov_svc_onco_core_prov_all_rbcs_nppes.parquet").as_posix()

con.execute(
    f"""
COPY prov_svc_onco_core_prov_all_rbcs_nppes_tbl
TO '{OUT_PARQUET}'
(FORMAT PARQUET);
"""
)
````

#### Table 1: provider_year_features (small, stable, best for EDA + clustering)

````python
con.execute("""
CREATE OR REPLACE TABLE provider_year_features AS
SELECT
  Rndrng_NPI,
  Year,

  -- provider identity (should be constant within NPI-Year)
  MAX(Rndrng_Prvdr_Type) AS provider_type,
  MAX(is_core_scope) AS is_core_scope,
  MAX(Rndrng_Prvdr_State_Abrvtn) AS state,
  MAX(Rndrng_Prvdr_Zip5) AS zip5,
  MAX(Rndrng_Prvdr_RUCA) AS ruca,
  MAX(Rndrng_Prvdr_RUCA_Desc) AS ruca_desc,

  -- provider-year totals (from provider summary)
  MAX(Tot_HCPCS_Cds) AS tot_hcpcs_cds,
  MAX(Tot_Benes_1) AS tot_benes,
  MAX(Tot_Srvcs_1) AS tot_srvcs,
  MAX(Tot_Sbmtd_Chrg) AS tot_sbmtd_chrg,
  MAX(Tot_Mdcr_Alowd_Amt) AS tot_mdcr_allowed_amt,
  MAX(Tot_Mdcr_Pymt_Amt) AS tot_mdcr_payment_amt,
  MAX(Tot_Mdcr_Stdzd_Amt) AS tot_mdcr_stdzd_amt,

  -- derived unit costs at provider-year level
  MAX(Tot_Mdcr_Stdzd_Amt) / NULLIF(MAX(Tot_Srvcs_1), 0) AS stdzd_amt_per_service,
  MAX(Tot_Mdcr_Pymt_Amt) / NULLIF(MAX(Tot_Srvcs_1), 0) AS payment_amt_per_service,
  MAX(Tot_Mdcr_Alowd_Amt) / NULLIF(MAX(Tot_Srvcs_1), 0) AS allowed_amt_per_service,

  -- drug vs medical totals (optional but useful)
  MAX(Drug_Tot_Srvcs) AS drug_tot_srvcs,
  MAX(Drug_Mdcr_Stdzd_Amt) AS drug_mdcr_stdzd_amt,
  MAX(Med_Tot_Srvcs) AS med_tot_srvcs,
  MAX(Med_Mdcr_Stdzd_Amt) AS med_mdcr_stdzd_amt,

  -- case-mix proxies
  MAX(Bene_Avg_Risk_Scre) AS bene_avg_risk_score,

  -- demographics counts (optional)
  MAX(Bene_Avg_Age) AS bene_avg_age,
  MAX(Bene_Feml_Cnt) AS bene_female_cnt,
  MAX(Bene_Male_Cnt) AS bene_male_cnt,
  MAX(Bene_Dual_Cnt) AS bene_dual_cnt,
  MAX(Bene_Ndual_Cnt) AS bene_nondual_cnt,

  -- chronic condition percentages (keep as features, even if they are BIGINT)
  MAX(Bene_CC_PH_Cancer6_V2_Pct) AS pct_cancer6,
  MAX(Bene_CC_PH_Diabetes_V2_Pct) AS pct_diabetes,
  MAX(Bene_CC_PH_CKD_V2_Pct) AS pct_ckd,
  MAX(Bene_CC_PH_COPD_V2_Pct) AS pct_copd,
  MAX(Bene_CC_PH_Hypertension_V2_Pct) AS pct_htn,

  -- experience proxy (from NPPES join)
  MAX(years_since_enumeration) AS years_since_enumeration

FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
GROUP BY 1, 2;
""")
````

````python
OUT = (DATA / "provider_year_features.parquet").as_posix()
con.execute(f"COPY provider_year_features TO '{OUT}' (FORMAT PARQUET);")
````

#### Table 2: provider_service_features (modeling table for regression benchmark)

````python
con.execute("""
CREATE OR REPLACE TABLE provider_service_features AS
SELECT
  Rndrng_NPI,
  Year,
  MAX(Rndrng_Prvdr_Type) AS provider_type,
  MAX(is_core_scope) AS is_core_scope,

  -- service grouping
  RBCS_FamNumb,
  MAX(RBCS_Family_Desc) AS rbcs_family_desc,
  MAX(RBCS_Cat) AS rbcs_cat,
  MAX(RBCS_Cat_Subcat) AS rbcs_cat_subcat,

  -- context
  Place_Of_Srvc,
  MAX(Rndrng_Prvdr_State_Abrvtn) AS state,
  MAX(Rndrng_Prvdr_Zip5) AS zip5,
  MAX(Rndrng_Prvdr_RUCA) AS ruca,

  -- volume and exposure
  SUM(Tot_Srvcs) AS services,
  SUM(Tot_Benes) AS benes,
  SUM(Tot_Bene_Day_Srvcs) AS bene_day_services,

  -- weighted average “price” targets (per service)
  SUM(Avg_Mdcr_Stdzd_Amt * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS stdzd_amt_per_service,
  SUM(Avg_Mdcr_Alowd_Amt * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS allowed_amt_per_service,
  SUM(Avg_Mdcr_Pymt_Amt * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS payment_amt_per_service,
  SUM(Avg_Sbmtd_Chrg * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS submitted_charge_per_service,

  -- provider-level features (repeated across service rows, so take MAX)
  MAX(Bene_Avg_Risk_Scre) AS bene_avg_risk_score,
  MAX(years_since_enumeration) AS years_since_enumeration,

  -- a few case-mix features (add more later if we want)
  MAX(Bene_CC_PH_Cancer6_V2_Pct) AS pct_cancer6,
  MAX(Bene_CC_PH_Diabetes_V2_Pct) AS pct_diabetes,
  MAX(Bene_CC_PH_CKD_V2_Pct) AS pct_ckd,
  MAX(Bene_CC_PH_COPD_V2_Pct) AS pct_copd,
  MAX(Bene_CC_PH_Hypertension_V2_Pct) AS pct_htn

FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
WHERE RBCS_FamNumb IS NOT NULL  -- keep only rows mapped to RBCS
GROUP BY 1,2,5,9;
""")
````

````python
OUT = (DATA / "provider_service_features.parquet").as_posix()
con.execute(f"COPY provider_service_features TO '{OUT}' (FORMAT PARQUET);")
````

### QC D. Reconciliation checks (to ensure we didn’t accidentally double-count)

6) Provider-year totals vs sum of provider-service “reconstructed totals”

````python
con.sql("""
WITH svc_recon AS (
  SELECT
    Rndrng_NPI,
    Year,
    SUM(stdzd_amt_per_service * services) AS recon_stdzd_amt,
    SUM(services) AS recon_services
  FROM provider_service_features
  GROUP BY 1,2
),
joined AS (
  SELECT
    y.Rndrng_NPI,
    y.Year,
    y.tot_mdcr_stdzd_amt,
    y.tot_srvcs,
    s.recon_stdzd_amt,
    s.recon_services,
    (s.recon_stdzd_amt / NULLIF(y.tot_mdcr_stdzd_amt, 0)) AS ratio_stdzd_amt,
    (s.recon_services / NULLIF(y.tot_srvcs, 0)) AS ratio_services
  FROM provider_year_features y
  LEFT JOIN svc_recon s
    ON y.Rndrng_NPI = s.Rndrng_NPI
   AND y.Year = s.Year
)
SELECT
  COUNT(*) AS n,
  AVG(CASE WHEN ratio_stdzd_amt BETWEEN 0.5 AND 1.5 THEN 1 ELSE 0 END) AS pct_ratio_stdzd_in_0_5_to_1_5,
  quantile_cont(ratio_stdzd_amt, 0.50) AS median_ratio_stdzd,
  quantile_cont(ratio_services, 0.50) AS median_ratio_services
FROM joined
WHERE tot_mdcr_stdzd_amt > 0 AND tot_srvcs > 0;
""").df()
````

## Creating the EDA dataset

````python
OUT = (DATA / "eda_dataset.parquet").as_posix()

con.execute("""
CREATE OR REPLACE TABLE eda_dataset AS
WITH base AS (
  SELECT
    -- keys
    ps.Rndrng_NPI,
    ps.Year,
    ps.RBCS_FamNumb,
    ps.Place_Of_Srvc,

    -- labels for readability (not for modeling)
    ps.rbcs_family_desc,
    ps.rbcs_cat,
    ps.rbcs_cat_subcat,

    -- provider context
    ps.provider_type,
    ps.is_core_scope,
    ps.state,
    ps.zip5,
    ps.ruca,

    -- derived RUCA bucket (simplifies plots and modeling later)
    CASE
      WHEN ps.ruca IS NULL THEN 'Unknown'
      WHEN ps.ruca <= 3 THEN 'Urban'
      WHEN ps.ruca <= 6 THEN 'Suburban'
      ELSE 'Rural'
    END AS ruca_bucket,

    -- exposure
    ps.services,
    ps.benes,
    ps.bene_day_services,

    -- observed “price” outcomes (per service)
    ps.stdzd_amt_per_service,
    ps.allowed_amt_per_service,
    ps.payment_amt_per_service,
    ps.submitted_charge_per_service,

    -- observed totals (useful for spend concentration plots)
    ps.stdzd_amt_per_service * ps.services AS stdzd_spend,
    ps.allowed_amt_per_service * ps.services AS allowed_spend,
    ps.payment_amt_per_service * ps.services AS payment_spend,
    ps.submitted_charge_per_service * ps.services AS submitted_spend,

    -- log transforms for EDA and later modeling
    ln(1 + ps.stdzd_amt_per_service) AS log_stdzd_amt_per_service,
    ln(1 + ps.services) AS log_services,
    ln(1 + ps.benes) AS log_benes,

    -- case-mix and experience
    ps.bene_avg_risk_score,
    ps.years_since_enumeration,

    -- condition percentage features (keep both raw and scaled)
    ps.pct_cancer6,
    ps.pct_diabetes,
    ps.pct_ckd,
    ps.pct_copd,
    ps.pct_htn,

    (CAST(ps.pct_cancer6 AS DOUBLE) / 100.0) AS p_cancer6,
    (CAST(ps.pct_diabetes AS DOUBLE) / 100.0) AS p_diabetes,
    (CAST(ps.pct_ckd AS DOUBLE) / 100.0) AS p_ckd,
    (CAST(ps.pct_copd AS DOUBLE) / 100.0) AS p_copd,
    (CAST(ps.pct_htn AS DOUBLE) / 100.0) AS p_htn,

    -- optional provider-year totals for context checks
    py.tot_benes,
    py.tot_srvcs,
    py.tot_mdcr_stdzd_amt,
    py.stdzd_amt_per_service AS provider_year_stdzd_amt_per_service

  FROM provider_service_features AS ps
  LEFT JOIN provider_year_features AS py
    ON ps.Rndrng_NPI = py.Rndrng_NPI
   AND ps.Year = py.Year

  WHERE ps.services >= 11
    AND ps.stdzd_amt_per_service IS NOT NULL
    AND ps.stdzd_amt_per_service >= 0
),
q AS (
  -- exact percentile (can swap to APPROX_QUANTILE for speed)
  SELECT quantile_cont(stdzd_amt_per_service, 0.99) AS q99
  FROM base
)
SELECT
  b.*,

  -- mirror pandas: is_top_1pct_stdzd_amt_per_service = stdzd_amt_per_service >= q99
  (b.stdzd_amt_per_service >= q.q99) AS is_top_1pct_stdzd_amt_per_service,

  -- mirror pd.cut(... right=False): left-inclusive, right-exclusive bins
  CASE
    WHEN b.services < 11 THEN '<11'
    WHEN b.services >= 11 AND b.services < 25 THEN '11-25'
    WHEN b.services >= 25 AND b.services < 50 THEN '25-50'
    WHEN b.services >= 50 AND b.services < 100 THEN '50-100'
    WHEN b.services >= 100 AND b.services < 500 THEN '100-500'
    ELSE '500+'
  END AS svc_bucket

FROM base b
CROSS JOIN q;
""")

con.execute(f"COPY eda_dataset TO '{OUT}' (FORMAT PARQUET);")
````

A sanity check that the top 1% flag is about right:

````python
con.sql(
    """
SELECT
  AVG(CASE WHEN is_top_1pct_stdzd_amt_per_service THEN 1 ELSE 0 END) AS pct_flagged
FROM eda_dataset;
"""
)
````

And verify bucket boundaries:

````python
con.sql(
    """
SELECT svc_bucket, MIN(services) AS min_s, MAX(services) AS max_s, COUNT(*) AS n
FROM eda_dataset
GROUP BY 1
ORDER BY 1;
"""
)
````

`tot_benes`: How many distinct beneficiaries did this provider see in this year across all services and places of service

`benes`: How many distinct beneficiaries received at least one service in this specific family and POS from this provider in this year?

So, we cannot sum `benes` and expect to get `tot_benes` for a specific provider-year pair. Let's demonstrate this:

````python
eda_df["Rndrng_NPI"].sample(2)
````

````python
eda_df["Year"].sample(1)
````

````python
eda_df.query("Rndrng_NPI == 1629379565 and Year == 2022")[["benes","tot_benes","Year", "Rndrng_NPI","Place_Of_Srvc","rbcs_family_desc"]].head()
````


- `tot_benes` = 354 means:
    - Provider 1629379565 had 354 unique Medicare beneficiaries in 2022 across all services and places of service that feed into our provider-year table.

- Row-level `benes` means:
    - Within each provider-year + family + place of service slice, how many unique beneficiaries had at least one service in that slice.

So in our sample rows:
- Office E&M Established, POS=O: 51 unique beneficiaries had at least one such service.
- IMRT, POS=O: 245 unique beneficiaries had at least one IMRT service.
- Office E&M New, POS=O: 66 unique beneficiaries had at least one new-patient E&M service.
- Conventional Radiation Treatment, POS=O: 327 unique beneficiaries had at least one conventional radiation service.
- Telephone Services, POS=O: 23 unique beneficiaries had at least one telephone service.

And crucially:
Those beneficiary sets can overlap heavily (likely they do here).
That is why we can see big `benes` values across multiple families while `tot_benes` stays 354.

Same idea applies to `services` vs `tot_srvcs`. 
Let's quickly check:

Interpretation:
- share ≈ 1 means our row table and provider-year totals align (same universe, no duplication).
- share < 1 usually means our row table is a subset (filters).
- share > 1 is a red flag for duplication or a universe mismatch where row table includes more than provider-year table.


````python
con.sql(
    """
SELECT
  SUM(services) AS sum_services_rows,
  MAX(tot_srvcs) AS tot_srvcs_py,
  SUM(services) * 1.0 / NULLIF(MAX(tot_srvcs), 0) AS share
FROM eda_dataset
WHERE Rndrng_NPI = 1629379565 AND Year = 2022;
"""
)
````

**What these numbers say:**

- `tot_srvcs_py` = `11817` means: provider `1629379565` delivered `11,817` total services in `2022` in the provider-year summary table’s universe.
- `sum_services_rows` = `11410` means: when we sum services across all rows we kept in `eda_dataset` for that provider-year, we capture `11,410` services.

So we are missing:
- 11817 - 11410 = 407 services
- Share captured = 11410 / 11817 = 0.9656 (about 96.6%)

Interpretation:

This is a good sign.

- Share < 1 means our row-level table is not “more” than the provider-year total. So it is not behaving like a duplication problem.
- Share close to 1 means our row-level table covers most of the provider-year services, but not all. That is consistent with our filters and/or scope restrictions.

What could explain the missing 3.4% (407 services)?
Most likely one of these, and our pipeline already suggests them:

1. We filtered to core oncology specialties (is_core_scope)

- If `tot_srvcs` was computed for the provider-year regardless of specialty restrictions, our EDA table would drop some services.

2. Service-family mapping coverage

- Some HCPCS lines might not land in our included RBCS families or might fall into categories we excluded upstream.
- Even “No RBCS Family” does not guarantee we captured every HCPCS, depending on how our family tables were built.

3. Place-of-service scope differences

- If our EDA rows only include certain POS codes (even implicitly, based on what exists in the service table we used), we might miss some POS contributions that still count in provider-year totals.

4. Other feasibility filters

- We filtered out null or negative `stdzd_amt_per_service` (and we require `services >= 11` at the row grain). Those filters can remove some rows that still contribute to the provider-year totals.

**One quick follow-up diagnostic:**

````python
con.sql(
    """
SELECT
  Place_Of_Srvc,
  SUM(services) AS sum_services,
  MAX(tot_srvcs) AS tot_srvcs_py,
  SUM(services) * 1.0 / NULLIF(MAX(tot_srvcs), 0) AS share
FROM eda_dataset
WHERE Rndrng_NPI = 1629379565 AND Year = 2022
GROUP BY 1
ORDER BY sum_services DESC;
"""
)
````

````python
con.sql(
    """
SELECT
  rbcs_family_desc,
  SUM(services) AS sum_services
FROM eda_dataset
WHERE Rndrng_NPI = 1629379565 AND Year = 2022
GROUP BY 1
ORDER BY sum_services DESC
LIMIT 30;
"""
)
````

We did the “within eda_dataset” breakdown correctly. The reason we still cannot see the missing 407 there is simple:
- Our POS table is computed from `eda_dataset`, and `eda_dataset` already only contains rows that passed our EDA filters (`ps.services >= 11`, `stdzd_amt_per_service IS NOT NULL`, `stdzd_amt_per_service >= 0`, plus whatever is upstream in provider_service_features).
- So the 407 `services` are not in `eda_dataset` at all. That is why they do not show up under any POS or any family in our POS and family tables.

Let's run this on `provider_service_features` (the table we SELECT FROM to build `eda_dataset`):

- If `services_that_would_enter_eda_dataset` = `11410` and `services_in_provider_service_features` = 11817, then the 407 are being dropped by the EDA filters.
- If `services_in_provider_service_features` is already 11410 (or close), then the 407 never made it into `provider_service_features` in the first place. That points to an upstream scope restriction (specialty filtering, family mapping, etc.).


**Step 1. Confirm the loss happens at the EDA filtering step (most likely)**

````python
con.sql(
    """
WITH base AS (
  SELECT *
  FROM provider_service_features
  WHERE Rndrng_NPI = 1629379565 AND Year = 2022
)
SELECT
  SUM(services) AS services_in_provider_service_features,

  SUM(CASE
        WHEN services >= 11
         AND stdzd_amt_per_service IS NOT NULL
         AND stdzd_amt_per_service >= 0
        THEN services ELSE 0 END) AS services_that_would_enter_eda_dataset,

  SUM(services) - SUM(CASE
        WHEN services >= 11
         AND stdzd_amt_per_service IS NOT NULL
         AND stdzd_amt_per_service >= 0
        THEN services ELSE 0 END) AS services_dropped_by_eda_filters
FROM base;
"""
)
````

**Step 2. Attribute the dropped services to specific reasons**

If `dropped_services_lt_11` is about 407, we have our answer: they are low-volume slices that we intentionally excluded for reliability.


````python
con.sql(
    """
WITH base AS (
  SELECT *
  FROM provider_service_features
  WHERE Rndrng_NPI = 1629379565 AND Year = 2022
)
SELECT
  SUM(CASE WHEN services < 11 THEN services ELSE 0 END) AS dropped_services_lt_11,

  SUM(CASE WHEN services >= 11 AND stdzd_amt_per_service IS NULL THEN services ELSE 0 END) AS dropped_services_stdzd_null,

  SUM(CASE WHEN services >= 11 AND stdzd_amt_per_service < 0 THEN services ELSE 0 END) AS dropped_services_stdzd_neg,

  SUM(services) AS total_services_in_base
FROM base;
"""
)
````

**Step 3. Show the exact dropped rows (so it’s not a mystery)**

This is the “where did they go” table.

This will show us the specific families and POS where those services live, and which rule excluded them.


````python
con.sql(
    """
SELECT
  Place_Of_Srvc,
  rbcs_family_desc,
  RBCS_FamNumb,
  services,
  benes,
  stdzd_amt_per_service,
  allowed_amt_per_service,
  payment_amt_per_service,
  submitted_charge_per_service
FROM provider_service_features
WHERE Rndrng_NPI = 1629379565 AND Year = 2022
  AND (
       services < 11
    OR stdzd_amt_per_service IS NULL
    OR stdzd_amt_per_service < 0
  )
ORDER BY services DESC
LIMIT 100;
"""
)
````

**Step 4. If the loss is upstream, compare against the raw service table**

Since Step 1 showed `provider_service_features` already below 11817, then we want to compare raw totals:


````python
con.sql(
    """
SELECT
  SUM(Tot_Srvcs) AS raw_sum_tot_srvcs
FROM prov_svc_all_prov_all
WHERE Rndrng_NPI = 1629379565 AND Year = 2022;
"""
)
````

````python
con.sql(
    """
SELECT
  SUM(services) AS services_in_provider_service_features
FROM provider_service_features
WHERE Rndrng_NPI = 1629379565 AND Year = 2022;
"""
)
````

````python
con.sql("""
WITH
psf AS (
  SELECT SUM(services) AS svc
  FROM provider_service_features
  WHERE Rndrng_NPI=1629379565 AND Year=2022
),
eda AS (
  SELECT SUM(services) AS svc
  FROM eda_dataset
  WHERE Rndrng_NPI=1629379565 AND Year=2022
),
raw AS (
  SELECT SUM(Tot_Srvcs) AS svc
  FROM prov_svc_all_prov_all
  WHERE Rndrng_NPI=1629379565 AND Year=2022
),
py AS (
  SELECT MAX(tot_srvcs) AS svc
  FROM provider_year_features
  WHERE Rndrng_NPI=1629379565 AND Year=2022
)
SELECT
  psf.svc AS services_provider_service_features,
  eda.svc AS services_eda_dataset,
  raw.svc AS services_raw_table_sum,
  py.svc  AS tot_srvcs_provider_year_features
FROM psf
CROSS JOIN eda
CROSS JOIN raw
CROSS JOIN py;
""")
````

Perfect. This is exactly the outcome we were trying to isolate.

What this proves (very clearly)

For this provider-year (NPI 1629379565, 2022):
- Raw service-line table sum = 11410
- provider_service_features sum = 11410
- eda_dataset sum = 11410

So our service-line universe is internally consistent end-to-end.

But:
- `provider_year_features.tot_srvcs` = 11817

So the “missing 407 services” are not missing from our service-line tables at all. They are only “missing” relative to the provider-year total table, which is using a broader definition of total services than what ended up in `prov_svc_all_prov_all` (and therefore in `provider_service_features` and `eda_dataset`).

Numerically:
- 11817 − 11410 = 407 services
- That is about 3.44% of the provider-year total (407 / 11817).

Why this happens (most likely)

`provider_year_features.tot_srvcs` is probably calculated from a different upstream table or rule set, for example:
- includes non-oncology specialties, non-core-scope, or non-included HCPCS/RBCS mappings
- includes rows we filtered out earlier in the pipeline (not at EDA time, but earlier during “oncology-core” construction)
- includes POS buckets or claim types not present in our final service-line extract
- includes services outside our RBCS mapping universe (even though we’re keeping “No RBCS Family” inside the service-line table).

We want to identify the universe used to compute `provider_year_features.tot_srvcs`. The fastest diagnostic is:

A) Find the exact table used to build `provider_year_features`
Where did `provider_year_features` come from in our notebook? Search for its CREATE TABLE statement and check which source table it aggregates.

B) Quantify the gap by computing provider-year totals from the same base as eda_dataset
If we want totals that reconcile by construction, compute a `“tot_srvcs_from_psf”` (or from the same service-line base) and carry that alongside the existing `tot_srvcs`.

**KEY NOTE: In EDA 8, we found that rows in the top 1% of `stdzd_amt_per_service` have a median service volume of 41. The median `stdzd_amt_per_service` among the top 1% is ~736.7, compared with ~70.3 for the bottom 99%.**

After this discovery, I decided to define tail flags and make them reusable. 

So let's create indicators once and use them everywhere:
- `is_top_1pct_stdzd_amt_per_service`
- and a service-volume (i.e., `services`) bucket: `svc_bucket`

````python
q99 = eda_df["stdzd_amt_per_service"].quantile(0.99)

eda_df = eda_df.assign(
    is_top_1pct_stdzd_amt_per_service = eda_df["stdzd_amt_per_service"] >= q99,
    svc_bucket = pd.cut(
        eda_df["services"],
        bins=[0, 11, 25, 50, 100, 500, np.inf],
        labels=["<11", "11-25", "25-50", "50-100", "100-500", "500+"],
        right=False
    )
)
````

Important realization: `“No RBCS Family”` showing up in `rbcs_family_desc` is not automatically wrong, but it is worth verifying what it corresponds to. It might be a real label in the RBCS file, or it might indicate some mapping fallback.

Quick check: If it’s truly mapped, RBCS_FamNumb should still be present and consistent.

````python
con.sql(
    """
SELECT
  rbcs_family_desc,
  COUNT(*) AS n,
  MIN(RBCS_FamNumb) AS min_fam,
  MAX(RBCS_FamNumb) AS max_fam
FROM provider_service_features
WHERE rbcs_family_desc = 'No RBCS Family'
GROUP BY 1;
"""
)
````

`"No RBCS Family"` looks like a real, intentional value in our RBCS mapping, not a join failure.

Because:

- We have **RBCS_FamNumb = ‘000’** for every one of those rows (min and max both `000`), and the description is consistently `"No RBCS Family"`.
- If this were a join miss, we would typically see `RBCS_FamNumb` as `NULL` (or missing) after the join, not a stable sentinel like `000`.
- We also previously saw `pct_missing_rbcs = 0.0`, which strongly suggests the taxonomy file contains entries for these HCPCS codes and is assigning them to a “none/other/unclassified” bucket.

So it is probably a legitimate “catch-all” category: “this HCPCS code is in the taxonomy file but not assigned to a meaningful family.” But we have to confirm this finding.

Quick checks: How big is it relative to the dataset?

````python
con.sql(
    """
SELECT
  COUNT(*) AS n_all,
  SUM(CASE WHEN RBCS_FamNumb='000' THEN 1 ELSE 0 END) AS n_000,
  1.0 * SUM(CASE WHEN RBCS_FamNumb='000' THEN 1 ELSE 0 END) / COUNT(*) AS pct_000
FROM provider_service_features;
"""
)
````

Quick checks: Does it have very different cost/exposure and spend?

````python
con.sql(
    """
SELECT
  CASE WHEN RBCS_FamNumb='000' THEN '000_NoFamily' ELSE 'OtherFamilies' END AS grp,
  COUNT(*) AS n_rows,
  AVG(services) AS avg_services,
  quantile_cont(log_stdzd_amt_per_service, 0.50) AS med_log_stdzd_amt_per_service,
  quantile_cont(log_stdzd_amt_per_service, 0.95) AS p95_log_stdzd_amt_per_service
FROM eda_dataset
GROUP BY 1;
"""
)
````

Let's calculate the % of total standardized spend coming from RBCS_FamNumb = ‘000':

1. of the meta dataset `prov_svc_onco_core_prov_all_rbcs_nppes_tbl`

````python
con.sql(
    """
SELECT
  SUM(Tot_Srvcs * Avg_Mdcr_Stdzd_Amt) AS total_stdzd_spend_all,
  SUM(CASE WHEN RBCS_FamNumb='000' THEN Tot_Srvcs * Avg_Mdcr_Stdzd_Amt ELSE 0 END) AS total_stdzd_spend_000,
  1.0 * SUM(CASE WHEN RBCS_FamNumb='000' THEN Tot_Srvcs * Avg_Mdcr_Stdzd_Amt ELSE 0 END)
      / NULLIF(SUM(Tot_Srvcs * Avg_Mdcr_Stdzd_Amt), 0) AS pct_stdzd_spend_000
FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl;
"""
)
````

2. of the EDA dataset `eda_dataset`:

````python
con.sql(
    """
SELECT
  SUM(stdzd_spend) AS total_stdzd_spend_all,
  SUM(CASE WHEN RBCS_FamNumb='000' THEN stdzd_spend ELSE 0 END) AS total_stdzd_spend_000,
  1.0 * SUM(CASE WHEN RBCS_FamNumb='000' THEN stdzd_spend ELSE 0 END)
      / NULLIF(SUM(stdzd_spend), 0) AS pct_stdzd_spend_000
FROM eda_dataset;
"""
)
````

This is exactly the kind of check that makes the story credible. Based on our results, I would treat RBCS_FamNumb = #### '000' as a real, meaningful segment and handle it explicitly.

**What these numbers are saying**

**Size**

- 000 is **7.9%** of rows (25,993 out of 329,571). That’s too big to ignore as a “quirk”, but small enough that it #### could distort things if it behaves differently.

**Exposure (services)**

- 000_NoFamily has **much higher average services** per row: ~12,234 vs ~2,836.
- That is a big structural difference. It suggests 000 rows are not “typical families”, they might be umbrella #### billing codes, very high volume codes, or catch-all codes that aggregate a lot of activity.

**Cost level (log standardized cost per service)**

- Median log cost is **lower** for 000 (3.38) vs others (4.29).
    - On the original scale, that’s roughly exp(3.38) ≈ 29 vs exp(4.29) ≈ 73 (this is approximate because we used log1p, but the conclusion holds).
- The **95th percentile is similar-ish** (6.14 vs 5.78). So the tail is not wildly different, but the center is.

So yes, I would call it meaningfully different. Even if the p95 is close, the median and the services volume difference are telling us this group is a different “kind” of row.





<br>

**Quick extra check before locking the decision**

The one thing I’d verify is whether 000 is concentrated in a specific place of service or provider type.

````python
con.sql(
    """
SELECT
  Place_Of_Srvc,
  provider_type,
  COUNT(*) AS n_rows,
  AVG(services) AS avg_services,
  quantile_cont(log_stdzd_amt_per_service, 0.50) AS med_log_stdzd
FROM eda_df
WHERE RBCS_FamNumb = '000'
GROUP BY 1,2
ORDER BY n_rows DESC;
"""
)
````

This table makes it very clear that **RBCS_FamNumb = '000'** is not just a harmless “misc” label. It behaves like multiple different sub-populations depending on place of service and provider type.

**What these result implies**

**1) The biggest driver is Place_Of_Srvc = O for HemOnc and MedOnc**

Look at these two rows:

- **O, Hematology-Oncology:** 10,741 rows, **avg_services ~ 22,765**, **median log cost ~ 1.16**
- **O, Medical Oncology:** 3,508 rows, **avg_services ~ 18,157**, **median log cost ~ 1.20**

That median log cost is extremely low relative to the rest of our dataset. These look like very high-volume, very low standardized cost-per-service rows, which is exactly the kind of pattern we get from “special” codes, aggregation-type codes, or something that is not comparable to a standard service family.

**2) The F rows look much more “normal”**

When Place_Of_Srvc = F, the median log costs are around ~4 to 6, and avg_services are tens, not tens of thousands. Example:

- **F, HemOnc:** avg_services ~ 79, med log ~ 4.10
- **F, RadOnc:** avg_services ~ 76, med log ~ 4.09
- **F, GynOnc:** avg_services ~ 39, med log ~ 6.24 (higher)

So 000 is not one thing. It’s at least two very different things depending on place of service.

**What to do with 000 now**

Two possible strategies:

**Option A (recommended): Keep 000, but model it explicitly with interactions**

Do all of this:

1. Add a flag
- is_no_rbcs_family = 1 if RBCS_FamNumb == '000' else 0
2. Add an interaction-type feature (either explicit or via model)
- For Ridge: we can explicitly create a feature like is_no_rbcs_family * is_place_O (or just rely on one-hot plus linear interaction if we build it).
- For XGBoost: we don’t need to manually create interactions. It will split on RBCS_FamNumb == '000' and Place_Of_Srvc naturally.
3. In EDA and reporting, always show metrics stratified by:
- is_no_rbcs_family and Place_Of_Srvc

This gives us the clean story: “000 codes behave differently, especially in outpatient settings. We keep them but let the model treat them differently.”

**Option B (cleanest benchmarking narrative): Separate-track modeling**

- **Main benchmark model:** train on RBCS_FamNumb != '000' only.
- **Secondary model:** train on RBCS_FamNumb == '000' only (and maybe even split by Place_Of_Srvc inside it).

This avoids mixing apples and oranges. It also prevents those huge-volume low-cost O rows from influencing the main model’s learned relationships.

**Which option fits our goal best?**

Since we want the headline metric to be **O/E standardized cost per service** and we want the benchmarking story to be clean, here’s the best alignment:

- If the report/dashboard is meant to be “benchmark real oncology service families,” then **Option B** is the cleanest.
- If the product goal is “give every provider-year-service row an expected cost,” then **Option A** is better.



<br>

**One more tiny QC that will make the decision obvious**

Let's check whether those crazy high avg_services rows also dominate total volume. If (O, HemOnc) and (O, MedOnc) are most of the services too, then separating 000 is usually the safest.

````python
con.sql(
    """
SELECT
  Place_Of_Srvc,
  provider_type,
  SUM(services) AS total_services,
  COUNT(*) AS n_rows,
  1.0 * SUM(services) / (SELECT SUM(services) FROM eda_df WHERE RBCS_FamNumb='000') AS pct_of_000_services
FROM eda_df
WHERE RBCS_FamNumb='000'
GROUP BY 1,2
ORDER BY total_services DESC;
"""
)
````

This confirms it very strongly.

**What this table says in plain English**

For the “No RBCS Family” bucket (RBCS_FamNumb = '000'):

- **Outpatient HemOnc (O, Hematology-Oncology)** accounts for **~76.9%** of *all* services in the 000 bucket.
- **Outpatient MedOnc (O, Medical Oncology)** accounts for **~20.0%**.
- Together, those two rows account for **~97%** of all 000 services.

Everything else is tiny by comparison.

So **RBCS_FamNumb='000'** is not just 7.9% of rows. It is a bucket whose *service volume* is overwhelmingly dominated by a very specific pattern (Outpatient + HemOnc/MedOnc). That means if we keep 000 in the main model without special handling, it can easily distort training, calibration, and the O/E story.

**Best recommendation**

Use **Option B (separate-track modeling)** as the default.

**Why Option B is the cleanest for our stated goal**

narrative usually assumes “we’re comparing like-with-like service families.” The 000 bucket is not behaving like a normal service family. It’s acting like a special aggregation or fallback code bucket with huge volume and very different cost behavior.

So:

1. **Main benchmark model (the one we show by default)**
- Train and score on **RBCS_FamNumb != '000'** only.
- This becomes our primary O/E story.
2. **Secondary model (optional add-on)**
- Train a separate model for **RBCS_FamNumb = '000'*, and honestly this could be restricted to just:
- (Place_Of_Srvc='O' AND provider_type IN ('Hematology-Oncology','Medical Oncology'))
- because that’s basically the whole 000 volume anyway.
- Or we can still include all provider types, but the key is keeping it separate from the main benchmark.

**What to do in the dataset**

Keep 000 rows in our data, but tag them and split the pipeline:

- model_track = 'main' if RBCS_FamNumb != '000'
- model_track = 'no_family' if RBCS_FamNumb = '000'

Then we build:

- model_dataset_main.parquet
- model_dataset_no_family.parquet (optional, can come later)



<br>

**One more quick QC that’s worth doing**

Check whether 000 is also concentrated in **specific HCPCS codes** (often a sign it’s a special code pattern):

If we see a small number of HCPCS codes dominating, that further supports “special bucket” behavior.

````python
con.sql(
    """
SELECT
  p.HCPCS_Cd,
  MAX(p.HCPCS_Desc) AS hcpcs_desc,
  COUNT(*) AS n_rows,
  SUM(p.Tot_Srvcs) AS total_services
FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl p
WHERE p.RBCS_FamNumb = '000'
GROUP BY 1
ORDER BY total_services DESC
LIMIT 20;
"""
)
````

It seems like most of the descriptions are injections. We should confirm this by running a regex to show what percentage of all **RBCS_FamNumb = '000'** are described as "injections" at any context of specific drug.

Fast check: “Injection” share by rows and by services

If pct_services_injection is very high (for example > 0.7), then the “000 bucket” is essentially “drug units/injections” in our data.

Slightly broader keyword net

Some injection-related codes might not literally contain the word “injection” (for example contrast materials). We can create a broader “drug or injectable” bucket:

This will tell us whether “injection” alone explains most of the services, or if it’s “injection + contrast + unclassified drug” etc.

````python
con.sql(
    """
WITH base AS (
  SELECT
    CASE
      WHEN lower(HCPCS_Desc) LIKE '%injection%' THEN 1
      ELSE 0
    END AS is_injection,
    Tot_Srvcs
  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE RBCS_FamNumb = '000'
)
SELECT
  COUNT(*) AS n_rows,
  SUM(is_injection) AS n_injection_rows,
  1.0 * SUM(is_injection) / COUNT(*) AS pct_rows_injection,

  SUM(Tot_Srvcs) AS total_services,
  SUM(CASE WHEN is_injection = 1 THEN Tot_Srvcs ELSE 0 END) AS injection_services,
  1.0 * SUM(CASE WHEN is_injection = 1 THEN Tot_Srvcs ELSE 0 END) / NULLIF(SUM(Tot_Srvcs), 0) AS pct_services_injection
FROM base;
"""
)
````

Bonus: Let's use `HCPCS_Drug_Ind` in our `prov_svc_onco_core_prov_all_rbcs_nppes_tbl`

Our big table appears to have `HCPCS_Drug_Ind`. If it’s reliable, it’s cleaner than regex:

**What this means for our modeling decision**

If we confirm that 000 is mostly drug-unit injection codes:

- It strengthens the case to **exclude RBCS_FamNumb='000' from the main service-family benchmark model** (our Option B main track).
- Then we can treat 000 as a **secondary “drug units” track** later, with its own modeling logic and careful interpretation.

````python
con.sql(
    """
WITH base AS (
  SELECT
    lower(HCPCS_Desc) AS d,
    Tot_Srvcs
  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE RBCS_FamNumb = '000'
),
tagged AS (
  SELECT
    CASE
      WHEN d LIKE '%injection%' THEN 'injection'
      WHEN d LIKE '%contrast%' THEN 'contrast'
      WHEN d LIKE '%unclassified drug%' OR d LIKE '%unclassified drugs%' THEN 'unclassified_drug'
      WHEN d LIKE '%chemotherap%' THEN 'chemo'
      WHEN d LIKE '%imaging%' OR d LIKE '%gadol%' OR d LIKE '%radiology%' THEN 'imaging_related'
      ELSE 'other'
    END AS bucket,
    Tot_Srvcs
  FROM base
)
SELECT
  bucket,
  COUNT(*) AS n_rows,
  SUM(Tot_Srvcs) AS total_services,
  1.0 * SUM(Tot_Srvcs) / (SELECT SUM(Tot_Srvcs) FROM tagged) AS pct_services
FROM tagged
GROUP BY 1
ORDER BY total_services DESC;
"""
)
````

````python
con.sql(
    """
SELECT
  HCPCS_Drug_Ind,
  COUNT(*) AS n_rows,
  SUM(Tot_Srvcs) AS total_services,
  1.0 * SUM(Tot_Srvcs) / (SELECT SUM(Tot_Srvcs)
                          FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
                          WHERE RBCS_FamNumb='000') AS pct_services
FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
WHERE RBCS_FamNumb='000'
GROUP BY 1
ORDER BY total_services DESC;
"""
)
````

It seems that most of the RBCS_FamNumb='000'  are actually related to injections.

These three checks line up perfectly, and they give us a very defensible conclusion.
What our results mean

1) “Injection” dominates by services (the important lens)

<br>

Only 53.4% of rows contain the word “injection”.
But those rows account for 91.3% of total services in the 000 bucket.
So the 000 bucket is not “half injections” in the sense that matters for modeling. It’s overwhelmingly injection volume.

<br>

2) Our broader keyword tagging confirms the same story

Injection: 91.3% of services
Contrast: 6.8% of services
Everything else combined is small.
This is very consistent with “drug units and related items” rather than standard “service families.”

<br>

3) The drug indicator seals it

<br>

HCPCS_Drug_Ind = 'Y' is 98.5% of services in 000.
That is huge. It means RBCS_FamNumb='000' is basically a drug-related bucket in our extracted oncology universe.
What we should do (clear recommendation)
Use the split-track approach:
Main benchmarking model (our portfolio headline)
Train and score ONLY on rows where:
RBCS_FamNumb IS NOT NULL
RBCS_FamNumb <> '000'
This is our clean “apples-to-apples service-family benchmarking engine.”
Secondary “drug unit” track (optional add-on)

<br>

Keep RBCS_FamNumb='000' rows separate.
If we include it later, frame it honestly as “drug unit and related codes (injections, contrast).”
It can be its own model or a descriptive module. It does not need to be part of the core benchmark story.
This aligns with our goal: make the primary narrative clean and interview-safe for Garner/Oscar.
How to describe this decision in our write-up

<br>

We can say something like:
“RBCS family ‘000’ represents non-standard family assignments in RBCS. In our oncology subset, it is overwhelmingly drug-related volume: 98.5% of services have HCPCS_Drug_Ind='Y', and 91.3% of services are codes whose description includes ‘injection’. Because drug unit billing behaves differently from service-family costs, we exclude RBCS ‘000’ from the main service-family benchmark model and treat it as a separate track.”
That sounds thoughtful, not like data cleaning hand-waving.

Let's now calculate within `RBCS_FamNumb='000'` what is the % of standardized spend from injection vs other

A) Text-based (“injection” in description), using the HCPCS table

````python
con.sql(
    """
WITH base AS (
  SELECT
    CASE WHEN lower(HCPCS_Desc) LIKE '%injection%' THEN 'injection' ELSE 'other' END AS bucket,
    Tot_Srvcs * Avg_Mdcr_Stdzd_Amt AS stdzd_spend
  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE RBCS_FamNumb='000'
)
SELECT
  bucket,
  COUNT(*) AS n_rows,
  SUM(stdzd_spend) AS total_stdzd_spend,
  1.0 * SUM(stdzd_spend) / NULLIF((SELECT SUM(stdzd_spend) FROM base), 0) AS pct_stdzd_spend
FROM base
GROUP BY 1
ORDER BY total_stdzd_spend DESC;
"""
)

````

B) Drug-indicator-based (often cleaner than text):

This uses `HCPCS_Drug_Ind` variable instead of relying on character matching.

````python
con.sql(
    """
WITH base AS (
  SELECT
    CASE WHEN HCPCS_Drug_Ind='Y' THEN 'drug_ind_Y' ELSE 'drug_ind_N' END AS bucket,
    Tot_Srvcs * Avg_Mdcr_Stdzd_Amt AS stdzd_spend
  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE RBCS_FamNumb='000'
)
SELECT
  bucket,
  COUNT(*) AS n_rows,
  SUM(stdzd_spend) AS total_stdzd_spend,
  1.0 * SUM(stdzd_spend) / NULLIF((SELECT SUM(stdzd_spend) FROM base), 0) AS pct_stdzd_spend
FROM base
GROUP BY 1
ORDER BY total_stdzd_spend DESC;
"""
)

````

**The interesting part: why “injection dominates services” but not “spend”**

Earlier, we found within 000:

- ~**98% of services** are “injection” (text match), and ~**98.6% of services** have `HCPCS_Drug_Ind='Y'`.

But now we see **only ~42% to 45% of spend** is injection or drug-indicated.

That usually means: **the injection/drug rows have massive unit counts (Tot_Srvcs) but low standardized amount per unit**, while the non-injection or non-drug rows have fewer units but higher standardized amounts per unit.

To make that relationship explicit, let's run this:

That will quantify the “lots of units, cheap per unit” vs “few units, expensive per unit” pattern.

````python
con.sql(
    """
WITH base AS (
  SELECT
    CASE WHEN lower(HCPCS_Desc) LIKE '%injection%' THEN 'injection' ELSE 'other' END AS bucket,
    Tot_Srvcs AS services,
    (Tot_Srvcs * Avg_Mdcr_Stdzd_Amt) AS stdzd_spend
  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE RBCS_FamNumb='000'
)
SELECT
  bucket,
  SUM(services) AS total_services,
  SUM(stdzd_spend) AS total_stdzd_spend,
  SUM(stdzd_spend) / NULLIF(SUM(services), 0) AS implied_stdzd_amt_per_service,
  1.0 * SUM(services) / NULLIF((SELECT SUM(services) FROM base),0) AS pct_services,
  1.0 * SUM(stdzd_spend) / NULLIF((SELECT SUM(stdzd_spend) FROM base),0) AS pct_spend
FROM base
GROUP BY 1
ORDER BY total_stdzd_spend DESC;
"""
)
````

What each column is telling us

Bucket totals
- Injection
    - total_services: 240,415,048.9 (98.13% of services)
    - total_stdzd_spend: $287,613,028.58 (45.36% of spend)
    - implied_stdzd_amt_per_service: $1.20 per service
- Other
    - total_services: 4,573,314.8 (1.87% of services)
    - total_stdzd_spend: $346,436,797.62 (54.64% of spend)
    - implied_stdzd_amt_per_service: $75.75 per service

The main takeaway

Within RBCS_FamNumb='000':
- Almost all of the service volume is “injection” (98.13%).
- But most of the standardized spend is “other” (54.64%).
- The reason is the per-service cost difference:
    - “Injection” averages ~$1.20 per unit
    - “Other” averages ~$75.75 per unit
    - That’s about a 63x difference (75.75 / 1.20 ≈ 63.3).




Those huge injection service counts are often because many drug HCPCS codes are billed in tiny units (mg, mcg, 1 mg units, etc.). To show that explicitly, let's pull the top injection HCPCS in 000 by services and look at their descriptions:

````python
con.sql(
    """
SELECT
  HCPCS_Cd,
  MAX(HCPCS_Desc) AS hcpcs_desc,
  SUM(Tot_Srvcs) AS total_services,
  SUM(Tot_Srvcs * Avg_Mdcr_Stdzd_Amt) AS total_stdzd_spend,
  SUM(Tot_Srvcs * Avg_Mdcr_Stdzd_Amt) / NULLIF(SUM(Tot_Srvcs),0) AS implied_amt_per_service
FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
WHERE RBCS_FamNumb='000'
  AND lower(HCPCS_Desc) LIKE '%injection%'
GROUP BY 1
ORDER BY total_services DESC
LIMIT 20;
"""
)
````

Let's quantify how much of injection volume comes from “per 1 mg / microgram / mcg / 0.1 ml” unit-style descriptions:

````python
con.sql(
    """
WITH inj AS (
  SELECT
    lower(HCPCS_Desc) AS d,
    Tot_Srvcs AS services
  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE RBCS_FamNumb='000'
    AND lower(HCPCS_Desc) LIKE '%injection%'
)
SELECT
  CASE
    WHEN d LIKE '% 1 mg%' OR d LIKE '%per 1 mg%' THEN 'per_1mg'
    WHEN d LIKE '%microgram%' OR d LIKE '% mcg%' THEN 'microgram_or_mcg'
    WHEN d LIKE '% 0.1 ml%' THEN 'per_0.1ml'
    ELSE 'other_injection_units'
  END AS unit_bucket,
  SUM(services) AS total_services,
  1.0 * SUM(services) / (SELECT SUM(services) FROM inj) AS pct_injection_services
FROM inj
GROUP BY 1
ORDER BY total_services DESC;
"""
)
````

**Adding counts per bin**

We need to know if 100+ has a small number of rows (unstable estimate) or tons (real pattern).

````python
eda_df.groupby("services_custom").agg(
    n_rows=("log_stdzd_amt_per_service","size"),
    var_y=("log_stdzd_amt_per_service", np.var),
    mean_y=("log_stdzd_amt_per_service","mean"),
    p95=("log_stdzd_amt_per_service", lambda s: np.quantile(s, 0.95))
)
````

**Split 100+ into sub-bins**

Keep our early bins, but break 100+ into a few more. For example:

- 101–250
- 251–1,000
- 1,001–10,000
- 10,000+

If the variance spike is really coming from, say, the “10k+” group, that is an important story.

````python
bins2 = [0, 11, 25, 50, 100, 250, 1000, 10000, np.inf]
labels2 = ["0-11","12-25","26-50","51-100","101-250","251-1k","1k-10k","10k+"]

eda_df["services_custom2"] = pd.cut(eda_df["services"], bins=bins2, labels=labels2, include_lowest=True)

sns.barplot(data=eda_df, x="services_custom2", y="log_stdzd_amt_per_service", estimator=np.var)
plt.xticks(rotation=30, ha="right")
plt.show()
````

**1) The “100+” bin is not “more reliable”. It’s a different regime.**

Look at these side by side:

- **51–100:** var ≈ **0.679**, IQR ≈ **0.780**, mean ≈ **4.47**, p95 ≈ **5.80**
- **100+:** var ≈ **1.874**, IQR ≈ **1.571**, mean ≈ **3.51**, p95 ≈ **5.04**

So 100+ is:

- **much more spread out** (variance and IQR both jump a lot),
- with a **lower mean** and **lower p95** than the smaller bins.

That means the huge variance is not just a few outliers. It’s telling us that 100+ rows are a heterogeneous mix (different families, different billing patterns, different settings), which is exactly what the sub-bin plot is showing.

**2) The split confirms the variance spike is concentrated in a middle-high range**

From the plot:

- 0–11 through 51–100 are all in the same ballpark (stable-ish).
- **101–250 and 251–1k** start drifting up a bit.
- **1k–10k** is the giant spike (the main problem).
- **10k+** drops again (still higher than low bins, but much lower than 1k–10k).

That is a classic “mixture” story: the 1k–10k segment likely contains specific high-volume drug administration patterns that have very different per-service cost behavior.

**Let's also use IQR as a robustness check**

Variance can be misleading. IQR tells us the “middle spread.”

````python
def iqr(s):
    return np.quantile(s, 0.75) - np.quantile(s, 0.25)

eda_df.groupby("services_custom").agg(
    n_rows=("log_stdzd_amt_per_service","size"),
    iqr_y=("log_stdzd_amt_per_service", iqr),
    var_y=("log_stdzd_amt_per_service", np.var)
)
````

````python
eda_df.groupby("services_custom2").agg(
    n_rows=("log_stdzd_amt_per_service","size"),
    iqr_y=("log_stdzd_amt_per_service", lambda s: np.quantile(s,0.75)-np.quantile(s,0.25)),
    var_y=("log_stdzd_amt_per_service", np.var)
    
)
````

**The key takeaway**

Our dataset is **row-heavy in the low and mid volumes**, but **service-heavy in the ultra-high volumes**.

- **10k+ is only 7.27% of rows** (23,947 rows)
- But it contains **84.17% of all services** (992M services)

So if we care about “what drives total oncology service volume”, the 10k+ group dominates. If we we about “typical provider-service records”, the smaller bins dominate.

This is exactly why we need a deliberate weighting strategy and why we should report results stratified by volume bin.

````python
tmp = (
    eda_df.groupby("services_custom2")
    .agg(
        n_rows=("services","size"),
        total_services=("services","sum")
    )
)
tmp["pct_rows"] = tmp["n_rows"] / tmp["n_rows"].sum()
tmp["pct_services"] = tmp["total_services"] / tmp["total_services"].sum()
tmp.sort_index()
````

**How to interpret the “Top families by total services” table**

That table is basically answering: **“What kinds of things are being done the most often (by count of services)?”**

- **Services_sum is volume**, not dollars.
- So it is totally possible for something to be huge in services but not huge in spend (cheap but frequent), and vice versa (expensive but not frequent).

**Why “No RBCS Family” is so high in services**

Given what we already proved earlier, **RBCS_FamNumb = ‘000’ is mostly injection-related and drug-indicated** (about ~91% of services have “injection” in the description, and ~98% of services are HCPCS_Drug_Ind = Y). That usually implies:

- It is likely a **real “unmapped/other” bucket**, not random noise, but it is also **not clinically interpretable as a real service family** like “Chemo agent” or “Radiation planning”.

So interpret it as: **“A big chunk of volume is in drug injection codes that are not mapped into a specific RBCS family.”**

That is not automatically “bad”. It just means we should treat it intentionally in EDA and modeling (more on that below).

**Where “No RBCS Family” fits here**

That matches our earlier diagnosis:

- Still, because volume is huge, it can become **material spend**.

**How to interpret the Pareto tables and plots (the key story)**

A Pareto answers: **“How concentrated is spend in a small number of families?”**

From our table **including** “No RBCS Family”:

- Top 1 family (Chemo agent) already gets us to **~23.5%** of spend.
- Top 5 gets us to **~63.4%**.
- Top 8 gets us to **~79.8%** (we’re basically at the 80% line).
- Top 12 gets us to **~90.1%**.

That is a very strong concentration story: **a small set of service families explains most total standardized spend**.

When we **exclude** RBCS_FamNumb = ‘000’, the curve gets even “cleaner” conceptually:

- Top 5 gets us to **~67.4%**.
- Top 7 gets us to **~78.5%**.
- Top 11 gets us to **~89.5%**.

So “000” is not driving the whole Pareto story. The story holds either way.

**What I would do with “No RBCS Family” going forward**

We have two defensible paths. Pick one and be consistent.

**Option 1 (recommended for clean storytelling)**

**Exclude FamNumb = ‘000’ from the “service mix and spend concentration” plots**, and explicitly label it as “unmapped drug injection codes” elsewhere.

Why:

- Those plots are meant to tell a clinical/operational story about families.
- “No RBCS Family” is not interpretable in the same way.

We can still keep it in the dataset for modeling if we want, but keep the family-level “mix” charts clean.

**Option 2 (keep it, but treat it as its own explicit bucket)**

Keep it in plots, but **rename it in charts** to something like:

- “Unmapped (000) drug injection codes”
    
    so readers do not think it is a meaningful service family.

**1C. Top 15 families by total standardized spend (the bottom 99% for standardized cost per service part of the `eda_df` dataset)**

**One small improvement to make this airtight:**

Let's add one more table that quantifies “tail enrichment” per family:

For each rbcs_family_desc, compute:

- share of spend in tail
- share of spend overall
- enrichment ratio = tail_share / overall_share

````python
# 1) totals for normalization
total_spend_all  = eda_df["stdzd_spend"].sum()
total_spend_tail = eda_df.loc[eda_df["is_top_1pct_stdzd_amt_per_service"], "stdzd_spend"].sum()

# 2) family-level spend (all vs tail)
fam_all = (eda_df.groupby("rbcs_family_desc", as_index=False)
           .agg(spend_all=("stdzd_spend", "sum")))

fam_tail = (eda_df.loc[eda_df["is_top_1pct_stdzd_amt_per_service"]]
            .groupby("rbcs_family_desc", as_index=False)
            .agg(spend_tail=("stdzd_spend", "sum")))

# 3) combine + compute shares and enrichment
tail_enrichment = (fam_all.merge(fam_tail, on="rbcs_family_desc", how="left")
                   .assign(
                       spend_tail=lambda d: d["spend_tail"].fillna(0.0),
                       overall_share=lambda d: d["spend_all"] / total_spend_all,
                       tail_share=lambda d: np.where(
                           total_spend_tail > 0, d["spend_tail"] / total_spend_tail, np.nan
                       ),
                       enrichment_ratio=lambda d: d["tail_share"] / d["overall_share"],
                       pct_spend_in_tail=lambda d: np.where(
                           d["spend_all"] > 0, d["spend_tail"] / d["spend_all"], np.nan
                       ),
                   )
                   .sort_values("enrichment_ratio", ascending=False))

# look at the most tail-enriched families
tail_enrichment.head(15)
````

The table above has, for each `rbcs_family_desc`:

- `overall_share` = “Out of all standardized spend in the full dataset, what fraction belongs to this family?”

- `tail_share` = “Out of standardized spend within the tail only (top 1% rows by `stdzd_amt_per_service`), what fraction belongs to this family?”

- `enrichment_ratio` = `tail_share / overall_share`
    - ">1" means the family is overrepresented in the tail.
    - "<1" means the family is underrepresented in the tail.

- `pct_spend_in_tail` = “Of this family’s total spend, what percent is in the tail?”

**A. The tail is not “spread around”. It is heavily concentrated in `PET- Oncology`.**

Look at `PET- Oncology`:

- `tail_share` = 0.580777 (about 58% of all tail spend)
- `overall_share` = 0.02233379 (about 2.23% of all spend)
- `enrichment_ratio` = 26.0 (about 26x overrepresented in the tail)
- `pct_spend_in_tail` = 0.450119 (about 45% of PET spend is in the tail)

*This tells a very clean story: the high-cost-per-service tail is largely a PET story.*

**B. Some families are “almost entirely tail”, but they are tiny overall.**

Examples:

- `Vascular Embolization` has `pct_spend_in_tail` ~ 0.99, enrichment ~57x
- `Arthroplasty Knee` and `Arthroplasty Hip` have `pct_spend_in_tail` ~ 0.96 and 0.94, enrichment ~55x

*This means: these families have very small total spend (spend_all is tiny compared to PET or IMRT), but when they show up, they tend to show up as extreme cost-per-service rows.*

These are “spiky” families. Great for interpretability, not necessarily major drivers of overall spend.

**C. The huge spend families are not tail-driven.**

Look at big baseline spend families:

- Conventional Radiation Treatment

    - `overall_share` ~ 9.56%
    - `tail_share` ~ 3.48%
    - `enrichment_ratio` ~ 0.36 (underrepresented in tail)
    - `pct_spend_in_tail` ~ 0.63%

- IMRT

- `overall_share` ~ 8.32%
- `tail_share` ~ 4.84%
- `enrichment_ratio` ~ 0.58
- `pct_spend_in_tail` ~ 1.0%

So our earlier EDA story that “these are major spend families” remains true, and it is not being driven by a few insane tail observations.

**D. “No RBCS Family” is mildly tail-enriched but not dominated by the tail.**

- `enrichment_ratio` ~ 2.44
- `pct_spend_in_tail` ~ 4.22%

So `'000'` is not the main tail culprit. It is somewhat overrepresented, but the tail is not primarily a “No RBCS” artifact.

**E. Colony Stimulating Factors are meaningfully tail-enriched.**

`enrichment_ratio` ~ 7.23
`pct_spend_in_tail` ~ 12.5%

So this family does have a real upper tail presence and could matter for modeling behavior, depending on what we are predicting and how we evaluate.

**Interpretation of the tail enrichment analysis**

- The top 1% cost-per-service tail is compositionally different from the full dataset.

- Tail spend is highly concentrated in PET- Oncology, which accounts for ~58% of tail spend while being ~2.2% of overall spend (about 26x enrichment).

- Several procedure families are almost entirely tail when present (embolization, arthroplasty, neurostimulator), but they contribute little to overall spend because their total spend is small.

- High-spend baseline families (chemo, IMRT, conventional RT, office E&M) are not tail-dominated, suggesting our primary spend concentration story is robust.

**Implications for modeling**

**A.We should keep the tail but evaluate it explicitly**

Our enrichement analysis makes the tail interpretable rather than "noise." This is why we should not blindly winsorize away signal.

We will
- train our model on `log_stdzd_amt_per_service`
- add an evaluation slice
    - performance on tail rows (`is_top_1pct_stdzd_amt_per_service == True`)
    - performance on non-tail rows (`is_top_1pct_stdzd_amt_per_service == False`)
This tells us if the model is failing exactly where the cost-outlier story lives. 

**B. Consider a two-stage modeling option (only if needed)**

Because PET is so tail-dominant, we might actually get a better model by doing:
1. Classify "tail vs non-tail", then
2. regress cost within each group

This is optional, and will be implemented only if needed. The enrichment analysis justifies this but it is not recommended right away. 

**C. Feature implications**
Because PET is so strong in tail composition, this reinforces that we want strong service-mix controls:
- `rbcs_family_desc` (or its higher-level category) MUST be included 
- `Place_Of_Service` also matters a lot for imaging vs facility pricing patterns.

**There seems to be a mismatch here!**

Let's investigate the following idea:

>CMS suppression (most common)
Many HCPCS-level rows are suppressed when beneficiary counts are small (privacy rule). Provider-year totals can still be reported even if many individual HCPCS lines are suppressed. Result: totals look big, visible HCPCS lines look incomplete.

**Check 1**. Compare provider summary counts vs observable HCPCS counts

````python
con.sql(
    """
SELECT
  MAX(Tot_Srvcs_1) AS tot_srvcs_summary,
  COUNT(*) AS n_hcpcs_rows_visible,
  SUM(Tot_Srvcs) AS sum_srvcs_visible
FROM prov_svc_onco_core_prov_all_rbcs_nppes
WHERE Rndrng_NPI=1629222450 AND Year=2022;

"""
)
````

**Check 2**. Is the table inherently “partial” for many providers

Pick a handful of NPIs and compute:

````python
con.sql(
    """
SELECT
  AVG(sum_srvcs_visible * 1.0 / NULLIF(tot_srvcs_summary,0)) AS avg_visible_share
FROM (
  SELECT
    Rndrng_NPI,
    Year,
    MAX(Tot_Srvcs_1) AS tot_srvcs_summary,
    SUM(Tot_Srvcs) AS sum_srvcs_visible
  FROM prov_svc_onco_core_prov_all_rbcs_nppes
  GROUP BY 1,2
);
"""
)
````

Let's investigate the following idea

>The table is filtered to an oncology-core or RBCS-mapped subset
The table includes filtering (“onco_core”, “all_rbcs”). If the provider-year totals are coming from the broader provider summary, but the HCPCS lines are only a subset (oncology-related, mapped, or kept after joins), we will also see gaps.

````python
con.execute(
    """
CREATE OR REPLACE VIEW prov_svc_all_prov_all AS
SELECT
  ps.*,
  p.* EXCLUDE (Rndrng_NPI, Year)
FROM prov_svc_all AS ps
LEFT JOIN prov_all AS p
  ON ps.Rndrng_NPI = p.Rndrng_NPI
 AND ps.Year = p.Year;
""")
````

***Two additional checks that would make this airtight***

**Check A**: Is the missing share correlated with small benes?

Suppression is usually more common when beneficiary counts are small.

If suppression is the driver, avg_visible_share should be lower in the small-benes buckets.

````python
con.sql(
    """
WITH py AS (
  SELECT
    Rndrng_NPI,
    Year,
    MAX(Tot_Benes_1) AS tot_benes_summary,
    MAX(Tot_Srvcs_1) AS tot_srvcs_summary,
    SUM(Tot_Srvcs)   AS sum_srvcs_visible
  FROM prov_svc_all_prov_all
  GROUP BY 1, 2
),
bucketed AS (
  SELECT
    CASE
      WHEN tot_benes_summary < 11 THEN '<11'
      WHEN tot_benes_summary < 50 THEN '11-49'
      WHEN tot_benes_summary < 100 THEN '50-99'
      ELSE '100+'
    END AS benes_bucket,
    sum_srvcs_visible * 1.0 / NULLIF(tot_srvcs_summary, 0) AS visible_share
  FROM py
)
SELECT
  benes_bucket,
  AVG(visible_share) AS avg_visible_share,
  APPROX_QUANTILE(visible_share, 0.10) AS p10,
  APPROX_QUANTILE(visible_share, 0.50) AS p50,
  APPROX_QUANTILE(visible_share, 0.90) AS p90,
  COUNT(*) AS n_provider_years
FROM bucketed
GROUP BY 1
ORDER BY
  CASE benes_bucket
    WHEN '<11' THEN 1
    WHEN '11-49' THEN 2
    WHEN '50-99' THEN 3
    ELSE 4
  END;
"""
)
````

It turns out the hypothesis "Suppression is usually more common when beneficiary counts are small." is false! 

**Check B**: Identify how extreme this provider-year is relative to others

Then check where 35/103 (~0.34) lands in that distribution. My guess is it will be in the “heavily suppressed” tail.

````python
con.sql(

"""
WITH py AS (
  SELECT
    Rndrng_NPI,
    Year,
    MAX(Tot_Srvcs_1) AS tot_srvcs_summary,
    SUM(Tot_Srvcs) AS sum_srvcs_visible
  FROM prov_svc_all_prov_all
  GROUP BY 1,2
),
ratio AS (
  SELECT
    *,
    sum_srvcs_visible * 1.0 / NULLIF(tot_srvcs_summary,0) AS visible_share
  FROM py
)
SELECT
  MIN(visible_share) AS min_share,
  APPROX_QUANTILE(visible_share, 0.05) AS p05,
  APPROX_QUANTILE(visible_share, 0.50) AS p50,
  APPROX_QUANTILE(visible_share, 0.95) AS p95,
  MAX(visible_share) AS max_share
FROM ratio;
"""
)
````

Since there is some level of suppression happening (more for the providers with less beneficiaries but more or less consistently across all beneficiary levels), we should check to see how much of this suppression is happening in our `eda_dataset`. 

Let’s now compute the provider-year visible_shares for the rows that actually enter`eda_dataset` after we apply filters (services ≥ 11, RBCS mapped), and see if the distribution changes. 

1) Provider-year visible share for the provider-years that enter eda_dataset

That produces one row per provider-year that actually makes it into `eda_dataset`, with `visible_share_eda`.

````python
con.sql(
    """
WITH py_eda AS (
  SELECT
    Rndrng_NPI,
    Year,
    SUM(services) AS sum_srvcs_in_eda,
    COUNT(*)      AS n_eda_rows
  FROM eda_dataset
  GROUP BY 1,2
),
py_totals AS (
  SELECT
    Rndrng_NPI,
    Year,
    tot_srvcs AS tot_srvcs_summary,
    tot_benes AS tot_benes_summary
  FROM provider_year_features
)
SELECT
  e.Rndrng_NPI,
  e.Year,
  e.sum_srvcs_in_eda,
  t.tot_srvcs_summary,
  e.sum_srvcs_in_eda * 1.0 / NULLIF(t.tot_srvcs_summary, 0) AS visible_share_eda,
  e.n_eda_rows,
  t.tot_benes_summary
FROM py_eda e
JOIN py_totals t
  ON e.Rndrng_NPI = t.Rndrng_NPI
 AND e.Year = t.Year
LIMIT 60;
"""
)
````

2) Summarize the distribution and compare to our earlier “HCPCS visible share” baseline

A) Distribution for the eda_dataset provider-years

````python
con.sql(
    """
WITH py_eda AS (
  SELECT
    Rndrng_NPI,
    Year,
    SUM(services) AS sum_srvcs_in_eda
  FROM eda_dataset
  GROUP BY 1,2
),
joined AS (
  SELECT
    e.Rndrng_NPI,
    e.Year,
    e.sum_srvcs_in_eda,
    p.tot_srvcs AS tot_srvcs_summary,
    e.sum_srvcs_in_eda * 1.0 / NULLIF(p.tot_srvcs, 0) AS visible_share_eda
  FROM py_eda e
  JOIN provider_year_features p
    ON e.Rndrng_NPI = p.Rndrng_NPI
   AND e.Year = p.Year
)
SELECT
  COUNT(*) AS n_provider_years,
  AVG(visible_share_eda) AS avg_share,
  APPROX_QUANTILE(visible_share_eda, 0.05) AS p05,
  APPROX_QUANTILE(visible_share_eda, 0.10) AS p10,
  APPROX_QUANTILE(visible_share_eda, 0.50) AS p50,
  APPROX_QUANTILE(visible_share_eda, 0.90) AS p90,
  APPROX_QUANTILE(visible_share_eda, 0.95) AS p95
FROM joined;
"""
)
````

B) Side-by-side comparison: “HCPCS visible share” vs “EDA share”, but on the same provider-years

This lets us see how much extra drop is coming from **our filters**, beyond CMS suppression.

````python
con.sql(
    """
WITH eda_py AS (
  SELECT
    Rndrng_NPI,
    Year,
    SUM(services) AS sum_srvcs_in_eda
  FROM eda_dataset
  GROUP BY 1,2
),
hcpcs_py AS (
  SELECT
    Rndrng_NPI,
    Year,
    MAX(Tot_Srvcs_1) AS tot_srvcs_summary,
    SUM(Tot_Srvcs)   AS sum_srvcs_visible_hcpcs
  FROM prov_svc_all_prov_all
  GROUP BY 1,2
),
joined AS (
  SELECT
    e.Rndrng_NPI,
    e.Year,
    h.tot_srvcs_summary,
    h.sum_srvcs_visible_hcpcs,
    e.sum_srvcs_in_eda,
    h.sum_srvcs_visible_hcpcs * 1.0 / NULLIF(h.tot_srvcs_summary, 0) AS visible_share_hcpcs,
    e.sum_srvcs_in_eda        * 1.0 / NULLIF(h.tot_srvcs_summary, 0) AS visible_share_eda
  FROM eda_py e
  JOIN hcpcs_py h
    ON e.Rndrng_NPI = h.Rndrng_NPI
   AND e.Year = h.Year
)
SELECT
  COUNT(*) AS n_provider_years,
  AVG(visible_share_hcpcs) AS avg_hcpcs_share,
  AVG(visible_share_eda)   AS avg_eda_share,
  APPROX_QUANTILE(visible_share_hcpcs, 0.50) AS p50_hcpcs,
  APPROX_QUANTILE(visible_share_eda,   0.50) AS p50_eda,
  APPROX_QUANTILE(visible_share_hcpcs, 0.90) AS p90_hcpcs,
  APPROX_QUANTILE(visible_share_eda,   0.90) AS p90_eda
FROM joined;
"""
)
````

1) The `eda_dataset` provider-years have a “pretty complete” picture, on average

From the `visible_share_eda` summary:

- **n = 56,804 provider-years**
- **avg_share = 0.7849**
- **median (p50) = 0.8913**
- **p90 = 0.9923**

Interpretation:

- For the provider-years that make it into `eda_dataset`, we typically observe **most** of their annual services. Median provider-year has about **89%** of services represented.
- But there is still a left tail. **p10 = 0.3469** means 10% of provider-years only have ~35% or less of their annual services represented in `eda_dataset`. So suppression (and/or the service-line inclusion rules) still causes partial visibility for some provider-years.

2) The side-by-side comparison shows us filters are not making visibility worse . If anything, they select “cleaner” provider-years

we got:

- **`avg_hcpcs_share` = 0.7449**
- **`avg_eda_share` = 0.7849**
- **`p50_hcpcs` = 0.8661**
- **`p50_eda` = 0.8913**
- **`p90_hcpcs` = 0.9782**
- **`p90_eda` = 0.9922**

Interpretation:

- If our `eda_dataset` filtering were removing a lot of services beyond CMS suppression, we'd expect **`avg_eda_share` < `avg_hcpcs_share`**.
- We see the opposite. `eda_share` is higher. That means the provider-years that survive into `eda_dataset` tend to be the ones where a larger portion of total services are observable at the service-line level.

**The two nuances not to be missed**

**Nuance A: We are conditioning on “provider-years that enter `eda_dataset`”**

This is selection. We are not saying suppression is gone overall. We are saying:

- “Among provider-years that pass RBCS mapping and services≥11 at the service-line grain, the visible share looks fine.”

That’s good, but it also implies our modeling dataset is biased toward provider-years with:

- higher volumes
- more non-suppressed lines
- more mappable services

That might be acceptable. We just want to be aware of it.

**Nuance B: Our `p95 = 1.0294` is a red flag (shares should not exceed 1)**

A share above 1 can happen because:

- totals and service-line counts are not strictly comparable (different definitions of “services”), or
- duplication/leakage in joins somewhere, or
- rounding plus approximate quantiles, but 1.03 is too high for just rounding.

This does not invalidate the general pattern, but we should sanity check it.

Let's run these two checks:

**Check 1**: how many provider-years have share > 1?

````python
con.sql(
    """
WITH py_eda AS (
  SELECT Rndrng_NPI, Year, SUM(services) AS sum_srvcs_in_eda
  FROM eda_dataset
  GROUP BY 1,2
),
joined AS (
  SELECT
    e.Rndrng_NPI,
    e.Year,
    e.sum_srvcs_in_eda,
    p.tot_srvcs AS tot_srvcs_summary,
    e.sum_srvcs_in_eda * 1.0 / NULLIF(p.tot_srvcs, 0) AS share
  FROM py_eda e
  JOIN provider_year_features p
    ON e.Rndrng_NPI=p.Rndrng_NPI AND e.Year=p.Year
)
SELECT
  SUM(CASE WHEN share > 1 THEN 1 ELSE 0 END) AS n_gt_1,
  MAX(share) AS max_share
FROM joined;
"""
)
````

**Check 2**: inspect the worst offenders

````python
con.sql(
    """
WITH py_eda AS (
  SELECT Rndrng_NPI, Year, SUM(services) AS sum_srvcs_in_eda
  FROM eda_dataset
  GROUP BY 1,2
),
joined AS (
  SELECT
    e.Rndrng_NPI,
    e.Year,
    e.sum_srvcs_in_eda,
    p.tot_srvcs AS tot_srvcs_summary,
    e.sum_srvcs_in_eda * 1.0 / NULLIF(p.tot_srvcs, 0) AS share
  FROM py_eda e
  JOIN provider_year_features p
    ON e.Rndrng_NPI=p.Rndrng_NPI AND e.Year=p.Year
)
SELECT *
FROM joined
WHERE share > 1
ORDER BY share DESC
LIMIT 20;
"""
)
````

The three key facts:

1. **share > 1 happens a lot**
- 4,198 out of 56,804 provider-years (about **7.4%**) have `sum_srvcs_in_eda` > `tot_srvcs_summary`.
2. **The maximum is exactly 2.0** (and several near 2)
- That pattern (exactly 2x or almost 2x) is a classic signature of **systematic duplication** of service-line rows before we aggregated into `eda_dataset`.
3. **`tot_srvcs_summary` comes from provider summary totals**, while `sum_srvcs_in_eda` comes from summing service-line rows.
- Provider totals are not “double-counted” by joins because we used `MAX(Tot_Srvcs_1)` in the provider-year table.
- Service-line sums can be inflated if the underlying service-line table has duplicate rows for the same “real” HCPCS line.

````python
con.sql(
    """
SELECT
  HCPCS_Cd,
  Place_Of_Srvc,
  COUNT(*) AS n_rows,
  SUM(Tot_Srvcs) AS sum_srvcs_across_rows,
  MAX(Tot_Srvcs) AS max_srvcs_single_row
FROM prov_svc_all_prov_all
WHERE Rndrng_NPI = 1912087271 AND Year = 2021
GROUP BY 1,2
HAVING COUNT(*) > 1
ORDER BY n_rows DESC, sum_srvcs_across_rows DESC
LIMIT 50;
"""
)
````

**Quick sanity check:**

we can verify that the problem is “duplication” and not something conceptual by comparing totals:

**Compare provider-year totals vs our EDA service sums**

If `max_diff` is very large (it likely will be for the big offenders), that reinforces this is not rounding. It is duplicated volume.

````python
con.sql(
    """
WITH eda_py AS (
  SELECT Rndrng_NPI, Year, SUM(services) AS sum_srvcs_in_eda
  FROM eda_dataset
  GROUP BY 1,2
)
SELECT
  COUNT(*) AS n,
  AVG(sum_srvcs_in_eda - p.tot_srvcs) AS avg_diff,
  MAX(sum_srvcs_in_eda - p.tot_srvcs) AS max_diff
FROM eda_py e
JOIN provider_year_features p
  ON e.Rndrng_NPI=p.Rndrng_NPI AND e.Year=p.Year
WHERE sum_srvcs_in_eda > p.tot_srvcs;
"""
)
````

Given what we just observed, the highest-probability source of duplication is our RBCS join (and much less likely the NPPES join). We should interrogate `prov_svc_onco_core_prov_all_rbcs` first, because that is exactly where a one-to-many mapping can silently replicate service-line rows.

Why the RBCS join is suspicious in our exact code 

```
LEFT JOIN rbcs AS r
  ON p.HCPCS_Cd = r.HCPCS_Cd
```

implicitly assumes **`rbcs.HCPCS_Cd` is unique** (one row per HCPCS code). If it is not unique, then one provider-service HCPCS row becomes 2+ rows after the join. That inflates:

- `SUM(Tot_Srvcs)` when we later aggregate
- `SUM(Avg_Mdcr_Stdzd_Amt * Tot_Srvcs)` (spend numerator)
- and therefore can distort service-line rollups and our later `SUM(services)` in `eda_dataset`

Our earlier “no duplicates” check was done on `prov_svc_all_prov_all`, which is pre-RBCS, so it is totally consistent to see no duplication there but still see duplication later.

**Step 1. Let's prove whether RBCS has duplicates per HCPCS code**

Let's run this first. It tells us whether our join can replicate rows.

````python
con.sql(
    """
SELECT
  COUNT(*) AS rbcs_rows,
  COUNT(DISTINCT HCPCS_Cd) AS distinct_hcpcs,
  COUNT(*) - COUNT(DISTINCT HCPCS_Cd) AS extra_rows_due_to_duplicates
FROM rbcs;
"""
)
````

Then let's identify the worst duplicate HCPCS codes:

If this returns anything, our join is one-to-many for those codes.

````python
con.sql(
    """
SELECT
  HCPCS_Cd,
  COUNT(*) AS n_rbcs_rows
FROM rbcs
GROUP BY 1
HAVING COUNT(*) > 1
ORDER BY n_rbcs_rows DESC
LIMIT 50;
"""
)
````

**Step 2. Measure row multiplication directly: before vs after RBCS join**

This is the most direct “did the join duplicate rows” check.

Pick the same offender provider-year we flagged (or any). For example: Rndrng_NPI=1912087271, Year=2021.

2A) Row counts

If `post_rbcs` has higher row count and higher `SUM(Tot_Srvcs)`, we have confirmed duplication is introduced by RBCS join.

````python
con.sql(
    """
SELECT
  'pre_rbcs' AS stage,
  COUNT(*) AS n_rows
FROM prov_svc_onco_core_prov_all
WHERE Rndrng_NPI=1912087271 AND Year=2021

UNION ALL

SELECT
  'post_rbcs' AS stage,
  COUNT(*) AS n_rows
FROM prov_svc_onco_core_prov_all_rbcs
WHERE Rndrng_NPI=1912087271 AND Year=2021;
"""
)
````

2B) Service totals

````python
con.sql(
    """
SELECT
  'pre_rbcs' AS stage,
  SUM(Tot_Srvcs) AS sum_tot_srvcs
FROM prov_svc_onco_core_prov_all
WHERE Rndrng_NPI=1912087271 AND Year=2021

UNION ALL

SELECT
  'post_rbcs' AS stage,
  SUM(Tot_Srvcs) AS sum_tot_srvcs
FROM prov_svc_onco_core_prov_all_rbcs
WHERE Rndrng_NPI=1912087271 AND Year=2021;
"""
)
````

**Step 3. Identify the exact lines getting duplicated after the RBCS join**

Let's do this by comparing the “natural key” of a provider-service line *before* RBCS, then checking how many RBCS matches it gets.

A good key for our provider-service HCPCS data is usually:

(Rndrng_NPI, Year, HCPCS_Cd, Place_Of_Srvc)

(If we want to be extra safe, we can include `HCPCS_Drug_Ind` too.)

````python
con.sql(
    """
SELECT
  p.HCPCS_Cd,
  p.Place_Of_Srvc,
  COUNT(*) AS n_post_join_rows,
  SUM(p.Tot_Srvcs) AS sum_srvcs_post_join,
  MAX(p.Tot_Srvcs) AS max_srvcs_single
FROM prov_svc_onco_core_prov_all_rbcs AS p
WHERE p.Rndrng_NPI=1912087271 AND p.Year=2021
GROUP BY 1,2
HAVING COUNT(*) > 1
ORDER BY n_post_join_rows DESC, sum_srvcs_post_join DESC
LIMIT 50;
"""
)
````

If we get results here but not in `prov_svc_all_prov_all`, it is basically a smoking gun.

Then let's inspect one duplicated HCPCS code and see what differs across the duplicated rows (RBCS columns):

````python
con.sql(
    """
SELECT
  Rndrng_NPI, Year, HCPCS_Cd, Place_Of_Srvc, Tot_Srvcs,
  RBCS_FamNumb,
  RBCS_Id,
  RBCS_Latest_Assignment,
  RBCS_Analysis_Start_Dt,
  RBCS_Analysis_End_Dt,
  Alt_Assignment_Method,
  RBCS_Id_Ever_Reassigned
FROM prov_svc_onco_core_prov_all_rbcs
WHERE Rndrng_NPI=1912087271 AND Year=2021
  AND HCPCS_Cd = '80502'
ORDER BY RBCS_Latest_Assignment DESC, RBCS_Analysis_Start_Dt DESC;
"""
)
````

#### **Investigating the standardized cost per service within `eda_df` quantiles:**

````python
def q1(s): return s.quantile(0.01)
def q5(s): return s.quantile(0.05)
def q90(s): return s.quantile(0.90)
def p95(s): return s.quantile(0.95)
def p99(s): return s.quantile(0.99)

eda_df[["stdzd_amt_per_service", "log_stdzd_amt_per_service"]].agg(
    ["mean", "min", q1, q5, "median", q90, p95, p99, "max"]
)
````

> *Notice: the `log_stdzd_amt_per_service` jumped from ~6.4 to ~10.6 when we go from 99th percentile to max value. This tells us that the top 1% slice has abnormally high log standardized cost per service. We must further characterize the top 1%.*

**One extra check before finalizing the EDA choice**

Because the `max` is so high, it’s worth confirming whether the top tail is driven by:

- tiny denominators (services=11 but big spend), or
- specific codes / POS, or
- data artifacts.

A quick diagnostic:

````python
eda_df.assign(
    is_top_1pct = eda_df["stdzd_amt_per_service"] >= eda_df["stdzd_amt_per_service"].quantile(0.99)
).groupby("is_top_1pct").agg(
    n=("stdzd_amt_per_service","size"),
    med_services=("services","median"),
    med_std=("stdzd_amt_per_service","median"),
)
````

**What this check tells us**

1) The extreme tail is real and it is meaningfully different, not just “a few giant values.”
- Bottom 99%: median `stdzd_amt_per_service` ≈ `70.3`
- Top 1%: median `stdzd_amt_per_service` ≈ `736.7`

That is roughly a 10x jump in the typical value inside the top 1%.

2) The top 1% tends to occur when service volume is smaller.
- Bottom 99%: median services = `142`
- Top 1%: median services = `41`

This pattern is consistent with “high cost per service often shows up in lower-volume lines,” which is plausible in claims data (rare codes, expensive drugs, certain settings). It also means those points can dominate plots on the raw scale even though they are only 1% of rows.

3) It argues against trimming for EDA.
If we trimmed (dropped) the top 1%, we would remove 3,184 rows that likely represent a real, high-cost segment. That can bias descriptive patterns. Winsorization keeps them but stops them from flattening our visualizations.

#### **Let's explicitly do tail analysis**

1. **Tail composition. Who is in the tail?**

Make the tail interpretable by breaking it down by:

- `rbcs_family_desc`
- `Place_Of_Srvc`
- `provider_type
- `state` (or `ruca_bucket`)
- `svc_bucket`

Do it in three ways because they answer different questions:

- **Row share**: “where do tail rows occur”
- **Service-weighted**: “where does tail volume sit”
- **Spend-weighted**: “where does tail money sit”

````python
tail = eda_df[eda_df["is_top_1pct_stdzd_amt_per_service"]]
base = eda_df

summary = (tail.groupby("rbcs_family_desc", as_index=False)
           .agg(n_rows=("rbcs_family_desc","size"),
                services=("services","sum"),
                spend=("stdzd_spend","sum"))
           .assign(pct_tail_rows=lambda d: d["n_rows"]/len(tail),
                   pct_all_rows=lambda d: d["n_rows"]/len(base),
                   pct_tail_services=lambda d: d["services"]/tail["services"].sum(),
                   pct_tail_spend=lambda d: d["spend"]/tail["stdzd_spend"].sum())
           .sort_values("spend", ascending=False)
           .head(20))

summary
````

````python
tail = eda_df[eda_df["is_top_1pct_stdzd_amt_per_service"]]
base = eda_df

summary = (tail.groupby("Place_Of_Srvc", as_index=False)
           .agg(n_rows=("Place_Of_Srvc","size"),
                services=("services","sum"),
                spend=("stdzd_spend","sum"))
           .assign(pct_tail_rows=lambda d: d["n_rows"]/len(tail),
                   pct_all_rows=lambda d: d["n_rows"]/len(base),
                   pct_tail_services=lambda d: d["services"]/tail["services"].sum(),
                   pct_tail_spend=lambda d: d["spend"]/tail["stdzd_spend"].sum())
           .sort_values("spend", ascending=False)
           .head(20))

summary
````

````python
tail = eda_df[eda_df["is_top_1pct_stdzd_amt_per_service"]]
base = eda_df

summary = (tail.groupby("provider_type", as_index=False)
           .agg(n_rows=("provider_type","size"),
                services=("services","sum"),
                spend=("stdzd_spend","sum"))
           .assign(pct_tail_rows=lambda d: d["n_rows"]/len(tail),
                   pct_all_rows=lambda d: d["n_rows"]/len(base),
                   pct_tail_services=lambda d: d["services"]/tail["services"].sum(),
                   pct_tail_spend=lambda d: d["spend"]/tail["stdzd_spend"].sum())
           .sort_values("spend", ascending=False)
           .head(20))

summary
````

B) Tail composition tables. Who shows up in the tail, by key categories

Let's do these as “tail vs overall” shares. Do not do a million plots.

Table B1: tail enrichment by `rbcs_family_desc` (we already did) We'll keep it. It is our main “tail is concentrated in these families” proof.

Add Table B2–B4: same idea but for POS, specialty, geography

These are cheap and very informative.

````python
def tail_enrichment_by(cat):
    d = eda_df.copy()
    tot_all = d["stdzd_spend"].sum()
    tot_tail = d.loc[d["is_top_1pct_stdzd_amt_per_service"], "stdzd_spend"].sum()

    all_ = d.groupby(cat, as_index=False).agg(spend_all=("stdzd_spend","sum"))
    tail_ = (d.loc[d["is_top_1pct_stdzd_amt_per_service"]]
               .groupby(cat, as_index=False)
               .agg(spend_tail=("stdzd_spend","sum")))

    out = (all_.merge(tail_, on=cat, how="left")
             .assign(
                 spend_tail=lambda x: x["spend_tail"].fillna(0.0),
                 overall_share=lambda x: x["spend_all"]/tot_all,
                 tail_share=lambda x: np.where(tot_tail>0, x["spend_tail"]/tot_tail, np.nan),
                 enrichment=lambda x: x["tail_share"]/x["overall_share"],
                 pct_spend_in_tail=lambda x: np.where(x["spend_all"]>0, x["spend_tail"]/x["spend_all"], np.nan),
             )
             .sort_values("enrichment", ascending=False))
    return out
````

These computed columns mean:

- `overall_share` = category’s share of all spend
- `tail_share` = category’s share of tail spend only
- `enrichment` = tail_share / overall_share
    - 1 means “overrepresented in the tail (by spend)”
    - < 1 means “underrepresented in the tail (by spend)”
- `pct_spend_in_tail` = fraction of that category’s spend that sits in the tail

````python
tail_enrichment_by("Place_Of_Srvc").head(20)
````

- `O` (Office): `enrichment` 1.056 (slightly overrepresented)
- `F` (Facility): `enrichment` 0.601 (underrepresented)

Takeaways:
- Tail spend is **a bit more office-weighted** than the overall spend mix.
- But this is not a huge effect. It is “directional,” not a dominant story.
- POS is still worth keeping as a feature.
- But POS alone is not explaining the tail.

## EDA 10. Modeling readiness summary (what EDA implies)

This section locks the modeling decisions implied by EDA. It serves as a contract for what will be trained, how it will be evaluated, and how “flagging” will be interpreted.

**Dataset grain (unit of analysis):** provider-year-RBCS family-place of service.

We explicitly separate:
- **Training inclusion rules** (coverage + stability)
- **Post-model flagging rules** (higher confidence, higher volume)

10.1 Objective, unit of prediction, and target

**Goal:** predict expected cost intensity per service for each provider-year-family-POS slice, adjusting for case-mix and context.

**Unit of prediction (row grain):**
- Provider (`Rndrng_NPI`)
- Year (`Year`)
- Service family (`RBCS_FamNumb` / `rbcs_family_desc`)
- Place of service (`Place_Of_Srvc`)

**Final modeling target:**
- `y = log1p(stdzd_amt_per_service)` (i.e., `log_stdzd_amt_per_service`)

**Why log target:**
- Raw `stdzd_amt_per_service` is heavy-tailed. Log transform stabilizes training while preserving rank and relative differences.

````python
target_decision = pd.DataFrame({
    "Decision": [
        "Outcome for reporting",
        "Modeling target",
        "Justification"
    ],
    "Value": [
        "stdzd_amt_per_service (raw, $/service)",
        "log1p(stdzd_amt_per_service) (stored as log_stdzd_amt_per_service)",
        "Heavy-tailed raw outcome. Log reduces tail dominance and improves stability."
    ]
})
target_decision
````

What it says
- Outcome for reporting: `stdzd_amt_per_service` (raw dollars per service)
- Modeling target: `log1p(stdzd_amt_per_service)` which we already store as `log_stdzd_amt_per_service`
- Justification: raw costs are heavy-tailed, log makes learning stable

How to interpret
- We are explicitly separating:
    - what stakeholders want to see and understand (raw dollars), from
    - what the model needs to learn well (log dollars).
- This is a strong and defensible stance because our EDA already proved:
    - max is huge (tens of thousands),
    - p99 is hundreds,
    - median is ~70.

Modeling implication
- Train the model on `log_stdzd_amt_per_service`.
- When we report results, convert back to dollars carefully:
- predicted log-cost -> `expm1(pred)` gives predicted $/service.
- residuals can be measured on log scale (stable) and optionally translated back to $ for interpretation.


10.2 Final training inclusion rules (EDA-aligned)

**Training inclusion filter:**
- `services >= 11`
- `stdzd_amt_per_service` is not null
- `stdzd_amt_per_service >= 0`

These rules prioritize reliability (avoid tiny denominators) and avoid invalid cost values.

````python
# Define training inclusion mask explicitly (even if eda_df is already filtered)
train_mask = (
    (eda_df["services"] >= 11) &
    (eda_df["stdzd_amt_per_service"].notna()) &
    (eda_df["stdzd_amt_per_service"] >= 0)
)

train_filter_summary = pd.DataFrame({
    "Metric": [
        "Rows in eda_df",
        "Rows meeting training filter",
        "Share kept",
        "Unique NPIs (all)",
        "Unique NPIs (kept)",
        "Years present (all)",
        "Years present (kept)",
    ],
    "Value": [
        len(eda_df),
        int(train_mask.sum()),
        float(train_mask.mean()),
        eda_df["Rndrng_NPI"].nunique(),
        eda_df.loc[train_mask, "Rndrng_NPI"].nunique(),
        sorted(eda_df["Year"].unique().tolist()),
        sorted(eda_df.loc[train_mask, "Year"].unique().tolist()),
    ]
})

train_filter_summary
````

The output
- `Rows in eda_df`: 318,322
- `Rows meeting training filter`: 318,322
- `Share kept`: 1.0
- `Unique NPIs`: 20,919
- `Years`: [2021, 2022, 2023]

How to interpret
- This confirms a key fact: our `eda_df` is already “training-ready” with respect to our inclusion rules.
- Our SQL filter `WHERE ps.services >= 11 AND ...` already enforced the train mask, so the explicit `train_mask` does not drop anything.

Modeling implication
- No additional inclusion filtering is needed in the modeling notebook.
- That is good because it prevents “silent changes” between EDA and modeling.

Narrative implication
“The modeling dataset equals the EDA dataset. No additional exclusions are introduced at modeling time.”


10.3 Tail policy (keep tail, do not winsorize for training)

**Key EDA finding:** the top 1% of `stdzd_amt_per_service` is not a handful of errors. It contains many rows and meaningful service volumes.

**Policy:**
- Do **not** winsorize labels for training.
- Use `log_stdzd_amt_per_service` as the modeling target to handle skew.
- Keep `is_top_1pct_stdzd_amt_per_service` as a diagnostic flag for slicing and interpretation.

**Visualization note:**
- Clipping at p99 is allowed **only for readability in plots**, not for training labels.

**10.3A: Tail footprint (rows, spend, services)**

````python
tail_footprint = (
    eda_df
    .assign(is_tail=eda_df["is_top_1pct_stdzd_amt_per_service"].astype(bool))
    .groupby("is_tail")
    .agg(
        n_rows=("services", "size"),
        total_services=("services", "sum"),
        total_spend=("stdzd_spend", "sum"),
        med_services=("services", "median"),
        med_cost=("stdzd_amt_per_service", "median"),
    )
    .assign(
        pct_rows=lambda d: d["n_rows"] / d["n_rows"].sum(),
        pct_services=lambda d: d["total_services"] / d["total_services"].sum(),
        pct_spend=lambda d: d["total_spend"] / d["total_spend"].sum(),
    )
)

tail_footprint
````

The `tail_footprint` output (key numbers)

- Tail flag is top 1% of `stdzd_amt_per_service` by construction.
- Non-tail:
    - `n_rows` = 315,138 (99.0%)
    - `total_services` ≈ 925,832,559.3
    - `total_spend` ≈ 1.488e10
    - `median services` = 142
    - `median cost` = ~70.3
    - `share of spend` = 98.27%
- Tail:
    - `n_rows` = 3,184 (1.0%)
    - `total_services` = 298,384
    - `total_spend` = 2.621e8
    - `median services` = 41
    - `median cost` = ~736.7
    - `share of spend` = 1.73%
    - `share of services` = 0.032%

How to interpret
This table answers three “so what?” questions:

A) Is the tail large in rows?
- It’s exactly 1% (3,184 rows). So it is not a tiny handful, but it is still a small slice.

B) Is the tail large in volume?
- No. Tail is only 0.032% of services.
- That is important: these are high cost per service, not high frequency.

C) Is the tail a major share of dollars?
- Surprisingly, also not huge: 1.73% of total spend.
- That means our dataset’s total spend is driven by the non-tail bulk, even though tail costs look extreme per service.

Modeling implication
- This strongly supports: no winsorization of labels needed.
- The tail is unlikely to dominate training loss when using log target, because:
    - it’s only 1% of rows,
    - it’s tiny in service volume,
    - log compresses extremes further.

Evaluation implication
- We should still slice metrics by tail vs non-tail, because the model might behave differently there.
- But tail is not our “main money story.” It is mainly a “rare high unit cost” story.


**10.3B: Tail explanation table (top 10 families by tail spend)**

````python
tail_tbl = (
    tail_enrichment_by("rbcs_family_desc")
    .sort_values("spend_tail", ascending=False)
    .loc[:, ["rbcs_family_desc", "tail_share", "pct_spend_in_tail", "enrichment", "spend_tail"]]
    .head(10)
    .assign(spend_tail_m=lambda d: d["spend_tail"] / 1e6)
    .assign(
        tail_share=lambda d: d["tail_share"].round(3),
        pct_spend_in_tail=lambda d: d["pct_spend_in_tail"].round(3),
        enrichment=lambda d: d["enrichment"].round(2),
        spend_tail_m=lambda d: d["spend_tail_m"].round(1),
    )
    .drop(columns=["spend_tail"])
)

tail_tbl
````

This is our “one slide” tail explanation.

Let’s interpret each column precisely:

-  `tail_share`:

“Of all dollars inside the tail, how much is contributed by this family?”

Example:
- PET-Oncology `tail_share` = 0.581
Means 58.1% of all tail spend is PET-Oncology.

This is a within-tail composition measure.

-  `pct_spend_in_tail`:

“Of this family’s total spend overall, what fraction lies in the tail?”

Example:
- Vascular Embolization `pct_spend_in_tail` = 0.991
Means 99.1% of all Vascular Embolization spend occurs in tail rows.

This is a within-family concentration measure.

- `enrichment`:

“How overrepresented is this category inside the tail compared to its overall prevalence?”

Formally:
- `enrichment` = tail_share / overall_share

Example:
- `PET-Oncology` `enrichment` = 26
Means `PET-Oncology` is 26x more concentrated inside the tail than expected if tail mirrored overall spend mix.

- `spend_tail_m`:

Raw dollars in the tail for that family in millions.

Example:
- `PET-Oncology` `spend_tail_m` = 152.3
Means `PET-Oncology` accounts for about $152.3M of tail spend.

What the `tail_tbl` tells us
- Tail spend is overwhelmingly driven by a few families, especially:
    - `PET-Oncology`
    - `Colony Stimulating Factors`
    - `No RBCS Family` (misc bucket)
- Some families have extreme `pct_spend_in_tail` (like `Vascular Embolization`) but small dollars overall.

Modeling implication
- The tail is not “random noise.” It is structured by service type.
- That is good news. It means categorical features like `rbcs_family_desc` and `Place_Of_Srvc` can help the model learn expected ranges.


10.4 Post-model flagging policy (separate from training)

Training includes rows with `services >= 11`.

For any downstream “flagging” use case (over-expected cost), we should apply stricter volume thresholds to ensure stability:
- **Candidate flagging threshold:** `services >= 50` (primary)
- **High-confidence slice:** `services >= 100` (sensitivity)

We will report model performance and flagging behavior across service-volume buckets.

**10.4A: Table A1 (tail flag x svc_bucket)**

````python
table_a1 = (
    eda_df
    .groupby(["is_top_1pct_stdzd_amt_per_service", "svc_bucket"], dropna=False, observed=True)
    .agg(
        n_rows=("services", "size"),
        total_services=("services", "sum"),
        total_spend=("stdzd_spend", "sum"),
        med_cost=("stdzd_amt_per_service", "median"),
        p95_cost=("stdzd_amt_per_service", lambda s: s.quantile(0.95)),
    )
    .assign(
        spend_per_service=lambda d: d["total_spend"] / d["total_services"],
        spend_per_row=lambda d: d["total_spend"] / d["n_rows"],
    )
    .assign(
        pct_services=lambda d: d["total_services"] / d["total_services"].sum(),
        pct_spend=lambda d: d["total_spend"] / d["total_spend"].sum(),
    )
    .sort_index()
)

table_a1
````

The `table_a1` table explains how volume interacts with tail membership.

Key patterns to notice

A) Most services are in non-tail 500+
- Non-tail 500+:
    - `total_services` ≈ 900,902,470
    - `pct_services` ≈ 0.973 (97.3% of all services)
    - `pct_spend` ≈ 0.846 (84.6% of all spend)
    - `spend_per_service` ≈ 14.23

Interpretation:
- The dataset is dominated by very high volume slices with low $/service.

B) Tail rows have much higher spend per service in every bucket

- Compare spend per service:
    - Non-tail `11-25`: ~124
    - Tail `11-25`: ~1103
    - Non-tail `100-500`: ~73
    - Tail `100-500`: ~775
    - Non-tail `500+`: ~14
    - Tail `500+`: ~732

Interpretation:

C) Tail spend is concentrated in mid-volume buckets, not the ultra-high bucket
- Look at tail rows:
    - Tail `500+`: only 83 rows
    - Tail `100-500`: 681 rows
    - Tail `11-25`: 1016 rows

Interpretation:
- The tail is mostly happening in low to mid volume slices, not the massive 500+ slices.

Modeling implication
- This supports our split policy:
    - train on `services >= 11` to have enough data
    - flagging at `services >= 50` or `>= 100` for stability
- It also suggests:
    - if we later do any “confidence scoring,” services volume should be part of it.

````python
flag_slices = pd.DataFrame({
    "Slice": ["services >= 11 (train)", "services >= 50 (flagging)", "services >= 100 (high confidence)"],
    "Rows": [
        int((eda_df["services"] >= 11).sum()),
        int((eda_df["services"] >= 50).sum()),
        int((eda_df["services"] >= 100).sum()),
    ],
    "Share of rows": [
        float((eda_df["services"] >= 11).mean()),
        float((eda_df["services"] >= 50).mean()),
        float((eda_df["services"] >= 100).mean()),
    ],
    "Share of spend": [
        float(eda_df.loc[eda_df["services"] >= 11, "stdzd_spend"].sum() / eda_df["stdzd_spend"].sum()),
        float(eda_df.loc[eda_df["services"] >= 50, "stdzd_spend"].sum() / eda_df["stdzd_spend"].sum()),
        float(eda_df.loc[eda_df["services"] >= 100, "stdzd_spend"].sum() / eda_df["stdzd_spend"].sum()),
    ],
})
flag_slices
````

**Missingness of features used in modeling**

````python
feature_missing = (
    model_df[cat_features + num_features + [target_col]]
    .isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .rename("pct_missing")
    .reset_index()
    .rename(columns={"index": "column"})
)
feature_missing
````

The `feature_missing` is the one we should pay attention to before training.

We have missingness in:
- `p_copd`: 4.59%
- `p_ckd`: 2.05%
- `p_diabetes`: 1.15%
- `p_cancer6`: 0.86%
- `years_since_enumeration`: 0.45%
- `p_htn`: 0.05%
Everything else: 0%

Interpret each variable’s missingness and what it implies

- `p_copd` (4.59% missing)
    - This is the highest missingness among our features.
    - Likely causes:
        - certain provider-year records lack the COPD percentage due to suppression, reporting rules, or merge gaps.
    - Modeling risk:
        - dropping rows would throw away ~4.6% of our data for one feature.
    - Recommended handling:
        - impute missing with a neutral value (commonly 0) plus a missingness indicator, or
        - impute with median and add indicator.
    - Why indicator matters:
        - missingness might correlate with provider type or geography, so the “missing” itself can carry signal.

- p_ckd (2.05% missing)
    - Similar logic, lower magnitude.
    - We should handle it the same way as p_copd for consistency.

- `p_diabetes` (1.15% missing)
    - Small but non-trivial.
    - Same treatment.

- `p_cancer6` (0.86% missing)
    - Small. Same treatment.
    - This one is especially sensitive conceptually in oncology, so do not silently drop rows.

- `years_since_enumeration` (0.45% missing)
    - This is “provider experience proxy.”
    - Missingness likely means:
        - NPI enumeration date missing upstream, or mapping failed.
    - Modeling risk:
        - leaving it missing can break some models.
    - Handling:
        - impute median and add missingness flag is the safest.

- `p_htn` (0.05% missing)
    - Very small. Still handle systematically (same imputation pattern).
    - Consistency matters. We do not want special-case logic for one feature.

- All categoricals have 0% missing
    - That is excellent. It means our slicing features are complete.
    - Especially important for:
        - `provider_type`, `Place_Of_Srvc`, `ruca_bucket`.

- `bene_avg_risk_score` has 0% missing
    - This is great because it is typically our primary adjustment feature.

Modeling implication
- We need a missingness strategy as part of Notebook 10 or the first modeling notebook.
- The best practice approach here is:

1.	For each numeric feature with missing:

- create `is_missing_<feature>` indicator

2.	Impute missing values:

- either 0 (for percentage fields) or median

3.	Keep the indicator in the model

This preserves rows, avoids bias from dropping, and allows missingness patterns to be learned.

**(`Rndrng_NPI`, `Year`, `Place_Of_Srvc`, `HCPCS_Cd`)**

That is provider-year-POS-HCPCS.

````python
OUT = (DATA / "eda_dataset_hcpcs.parquet").as_posix()

con.execute("""
CREATE OR REPLACE TABLE eda_dataset_hcpcs AS
WITH hcpcs_base AS (
  SELECT
    Rndrng_NPI,
    Year,
    Place_Of_Srvc,
    HCPCS_Cd,

    MAX(HCPCS_Desc) AS hcpcs_desc,
    MAX(HCPCS_Drug_Ind) AS hcpcs_drug_ind,

    MAX(RBCS_FamNumb) AS rbcs_famnumb,
    MAX(RBCS_Family_Desc) AS rbcs_family_desc,
    MAX(RBCS_Cat) AS rbcs_cat,
    MAX(RBCS_Cat_Subcat) AS rbcs_cat_subcat,

    MAX(Rndrng_Prvdr_Type) AS provider_type,
    MAX(is_core_scope) AS is_core_scope,
    MAX(Rndrng_Prvdr_State_Abrvtn) AS state,
    MAX(Rndrng_Prvdr_Zip5) AS zip5,
    MAX(Rndrng_Prvdr_RUCA) AS ruca,
    MAX(Rndrng_Prvdr_RUCA_Desc) AS ruca_desc,

    SUM(Tot_Srvcs) AS services,
    SUM(Tot_Benes) AS benes,
    SUM(Tot_Bene_Day_Srvcs) AS bene_day_services,

    SUM(Avg_Mdcr_Stdzd_Amt * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS avg_mdcr_stdzd_amt,
    SUM(Avg_Mdcr_Alowd_Amt * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS avg_mdcr_allowed_amt,
    SUM(Avg_Mdcr_Pymt_Amt  * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS avg_mdcr_payment_amt,
    SUM(Avg_Sbmtd_Chrg     * Tot_Srvcs) / NULLIF(SUM(Tot_Srvcs), 0) AS avg_sbmtd_chrg,

    SUM(Avg_Mdcr_Stdzd_Amt * Tot_Srvcs) AS stdzd_spend,
    SUM(Avg_Mdcr_Alowd_Amt * Tot_Srvcs) AS allowed_spend,
    SUM(Avg_Mdcr_Pymt_Amt  * Tot_Srvcs) AS payment_spend,
    SUM(Avg_Sbmtd_Chrg     * Tot_Srvcs) AS submitted_spend,

    MAX(Bene_Avg_Risk_Scre) AS bene_avg_risk_score,
    MAX(years_since_enumeration) AS years_since_enumeration,

    MAX(Bene_CC_PH_Cancer6_V2_Pct) AS pct_cancer6,
    MAX(Bene_CC_PH_Diabetes_V2_Pct) AS pct_diabetes,
    MAX(Bene_CC_PH_CKD_V2_Pct) AS pct_ckd,
    MAX(Bene_CC_PH_COPD_V2_Pct) AS pct_copd,
    MAX(Bene_CC_PH_Hypertension_V2_Pct) AS pct_htn

  FROM prov_svc_onco_core_prov_all_rbcs_nppes_tbl
  WHERE HCPCS_Cd IS NOT NULL
    AND RBCS_FamNumb IS NOT NULL
  GROUP BY 1,2,3,4
),
hcpcs_enriched AS (
  SELECT
    b.*,
    CASE
      WHEN b.ruca IS NULL THEN 'Unknown'
      WHEN b.ruca <= 3 THEN 'Urban'
      WHEN b.ruca <= 6 THEN 'Suburban'
      ELSE 'Rural'
    END AS ruca_bucket,

    ln(1 + b.avg_mdcr_stdzd_amt) AS log_avg_mdcr_stdzd_amt,
    ln(1 + b.stdzd_spend) AS log_stdzd_spend,
    ln(1 + b.services) AS log_services,
    ln(1 + b.benes) AS log_benes,

    (CAST(b.pct_cancer6 AS DOUBLE) / 100.0) AS p_cancer6,
    (CAST(b.pct_diabetes AS DOUBLE) / 100.0) AS p_diabetes,
    (CAST(b.pct_ckd AS DOUBLE) / 100.0) AS p_ckd,
    (CAST(b.pct_copd AS DOUBLE) / 100.0) AS p_copd,
    (CAST(b.pct_htn AS DOUBLE) / 100.0) AS p_htn

  FROM hcpcs_base b
  WHERE b.services >= 11
    AND b.avg_mdcr_stdzd_amt IS NOT NULL
    AND b.avg_mdcr_stdzd_amt >= 0
),
hcpcs_with_lags AS (
  SELECT
    e.*,
    LAG(e.avg_mdcr_stdzd_amt) OVER (
      PARTITION BY e.Rndrng_NPI, e.Place_Of_Srvc, e.HCPCS_Cd
      ORDER BY e.Year
    ) AS lag1_avg_amt,
    LAG(e.services) OVER (
      PARTITION BY e.Rndrng_NPI, e.Place_Of_Srvc, e.HCPCS_Cd
      ORDER BY e.Year
    ) AS lag1_tot_srvcs,
    LAG(e.stdzd_spend) OVER (
      PARTITION BY e.Rndrng_NPI, e.Place_Of_Srvc, e.HCPCS_Cd
      ORDER BY e.Year
    ) AS lag1_spend
  FROM hcpcs_enriched e
),
q AS (
  SELECT
    Year,
    quantile_cont(avg_mdcr_stdzd_amt, 0.99) AS q99
  FROM hcpcs_with_lags
  GROUP BY 1
)
SELECT
  h.*,
  (h.avg_mdcr_stdzd_amt >= q.q99) AS is_top_1pct_avg_mdcr_stdzd_amt,
  CASE
    WHEN h.services < 11 THEN '<11'
    WHEN h.services >= 11 AND h.services < 25 THEN '11-25'
    WHEN h.services >= 25 AND h.services < 50 THEN '25-50'
    WHEN h.services >= 50 AND h.services < 100 THEN '50-100'
    WHEN h.services >= 100 AND h.services < 500 THEN '100-500'
    ELSE '500+'
  END AS svc_bucket
FROM hcpcs_with_lags h
JOIN q USING (Year);
""")

con.execute(f"COPY eda_dataset_hcpcs TO '{OUT}' (FORMAT PARQUET);")
````

````python
con.sql("""
WITH q AS (
  SELECT
    Year,
    quantile_cont(avg_mdcr_stdzd_amt, 0.99) AS q99
  FROM eda_dataset_hcpcs
  GROUP BY 1
)
SELECT * FROM q ORDER BY Year;
""")
````

````python
con.sql("""
WITH q AS (
  SELECT
    Year,
    quantile_cont(avg_mdcr_stdzd_amt, 0.99) AS q99
  FROM eda_dataset_hcpcs
  GROUP BY 1
),
x AS (
  SELECT e.Year, e.avg_mdcr_stdzd_amt, q.q99
  FROM eda_dataset_hcpcs e
  JOIN q USING (Year)
)
SELECT
  Year,
  COUNT(*) AS n,
  SUM(CASE WHEN avg_mdcr_stdzd_amt >= q99 THEN 1 ELSE 0 END) AS n_ge,
  SUM(CASE WHEN avg_mdcr_stdzd_amt >  q99 THEN 1 ELSE 0 END) AS n_gt,
  n_ge::DOUBLE / COUNT(*) AS rate_ge,
  n_gt::DOUBLE / COUNT(*) AS rate_gt
FROM x
GROUP BY 1
ORDER BY 1;
""")
````

````python
con.sql("""
WITH q AS (
  SELECT Year, quantile_cont(avg_mdcr_stdzd_amt, 0.99) AS q99
  FROM eda_dataset_hcpcs
  GROUP BY 1
),
x AS (
  SELECT e.Year, e.avg_mdcr_stdzd_amt, q.q99
  FROM eda_dataset_hcpcs e
  JOIN q USING (Year)
)
SELECT
  Year,
  SUM(CASE WHEN avg_mdcr_stdzd_amt = q99 THEN 1 ELSE 0 END) AS n_eq_q99,
  SUM(CASE WHEN avg_mdcr_stdzd_amt > q99 THEN 1 ELSE 0 END) AS n_gt_q99,
  COUNT(*) AS n
FROM x
GROUP BY 1
ORDER BY 1;
""")
````

````python
# ============================================================
# HCPCS-level EDA Appendix (post-final-model, single source of truth)
# Source of truth: artifacts/eval_universe/eval_scored_DG_V3.parquet
# Row grain: (Rndrng_NPI, HCPCS_Cd, Place_Of_Srvc, Year)
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# 0) Load "single source of truth"
# ----------------------------
REPO_ROOT = Path.cwd()  # assuming notebook run from repo root; adjust if needed
EVAL_PATH = REPO_ROOT / "artifacts" / "eval_universe" / "eval_scored_DG_V3.parquet"
HCPCS_PATH = REPO_ROOT / "data" / "eda_dataset_hcpcs.parquet"

if EVAL_PATH.exists():
    SOURCE_NAME = "eval_scored_DG_V3.parquet"
    df = pd.read_parquet(EVAL_PATH)
elif HCPCS_PATH.exists():
    SOURCE_NAME = "eda_dataset_hcpcs.parquet"
    df = pd.read_parquet(HCPCS_PATH)
else:
    raise FileNotFoundError(
        f"Could not find either:\n- {EVAL_PATH}\n- {HCPCS_PATH}\n"
        "Run from repo root or update REPO_ROOT."
    )

print(f"[Loaded] {SOURCE_NAME}")
print("Columns:", list(df.columns)[:25], "...")


# ----------------------------
# Helpers
# ----------------------------
GRAIN_COLS = ["Rndrng_NPI", "HCPCS_Cd", "Place_Of_Srvc", "Year"]

def _require_cols(d: pd.DataFrame, cols: list[str], name: str):
    missing = [c for c in cols if c not in d.columns]
    if missing:
        raise ValueError(f"Missing required columns for {name}: {missing}")

def _fmt_int(x):
    try:
        return f"{int(x):,}"
    except Exception:
        return str(x)

def _safe_num(s):
    return pd.to_numeric(s, errors="coerce")

_require_cols(df, GRAIN_COLS, "grain confirmation")

# Ensure expected dtypes (safe casts)
df["Rndrng_NPI"] = df["Rndrng_NPI"].astype(str)
df["HCPCS_Cd"] = df["HCPCS_Cd"].astype(str)
df["Place_Of_Srvc"] = df["Place_Of_Srvc"].astype(str)
df["Year"] = _safe_num(df["Year"]).astype("Int64")

# Optional commonly used fields (don’t hard fail)
if "services" in df.columns:
    df["services"] = _safe_num(df["services"]).fillna(0)
if "benes" in df.columns:
    df["benes"] = _safe_num(df["benes"]).fillna(0)
if "state" in df.columns:
    df["state"] = df["state"].astype(str)
````

## ✅ A1) Grain confirmation (HCPCS-level dataset)

### What this table is proving
This check validates the **unit of analysis (row grain)** for the HCPCS-level dataset we use downstream for benchmarking, anomaly surfacing, and tiering.

**Declared row grain:**  
`(Rndrng_NPI, HCPCS_Cd, Place_Of_Srvc, Year)`

### Output summary (what we have)
- **Total rows:** `1,210,275`  
- **Unique providers (NPI):** `21,591`  
- **Unique HCPCS codes:** `1,742`  
- **Unique place of service (POS):** `2`  
- **Unique years:** `4`  

### Key interpretation
✅ **Duplicate rows at declared grain: `0`**  
This means every row is a **unique** `(NPI, HCPCS, POS, Year)` record. There are no accidental duplicates that would:
- double-count utilization or spending,
- distort model training/evaluation,
- inflate anomaly counts,
- or bias provider-level rollups.

### Why this matters downstream
This grain is exactly what we need because it aligns with the core benchmarking question:

> “For a given provider, procedure code (HCPCS), setting (POS), and year, how does observed cost compare to expected cost?”

This is also the grain at which:
- the regression model generates **expected_cost**,
- we compute **residual**, **oe_ratio**, and **log_oe**,
- and we define row-level anomalies in the worklists (ANOM.2 family).

### A subtle but important implication (provider counts)
We have **21,591 unique NPIs**, but later we sometimes work at a slightly different “provider profile” grain:
- **Provider profile grain:** `(Rndrng_NPI, provider_type, state)`

A single NPI can appear in multiple provider_type/state combinations in the dataset (e.g., taxonomy changes, multi-state billing), so **provider profiles** can exceed unique NPIs. That is expected and does not contradict the row grain.

### Takeaway
This dataset is **cleanly keyed** at the HCPCS benchmarking grain, which makes all downstream modeling and anomaly logic **auditable and reproducible**.

````python
# ============================================================
# A2) Time coverage
#   - plots: rows/year, providers/year, total services/year
#   - table: year min/max, share of rows by year
# ============================================================
time_tbl = (
    df.dropna(subset=["Year"])
      .groupby("Year", dropna=False)
      .agg(
          n_rows=("Year", "size"),
          n_providers=("Rndrng_NPI", "nunique"),
          total_services=("services", "sum") if "services" in df.columns else ("Year", "size"),
          total_benes=("benes", "sum") if "benes" in df.columns else ("Year", "size"),
      )
      .reset_index()
      .sort_values("Year")
)

time_tbl["row_share_pct"] = 100 * time_tbl["n_rows"] / max(time_tbl["n_rows"].sum(), 1)
year_min = int(time_tbl["Year"].min())
year_max = int(time_tbl["Year"].max())

print(f"Year range: {year_min} to {year_max}")
display(time_tbl)

# Plot: rows per year
fig = plt.figure(figsize=(8, 4))
ax = fig.add_subplot(111)
ax.plot(time_tbl["Year"].astype(int), time_tbl["n_rows"].astype(float), marker="o")
ax.set_title("Rows per Year")
ax.set_xlabel("Year")
ax.set_ylabel("Rows")
ax.set_ylim(0,350000)
plt.tight_layout()
plt.show()

# Plot: providers per year
fig = plt.figure(figsize=(8, 4))
ax = fig.add_subplot(111)
ax.plot(time_tbl["Year"].astype(int), time_tbl["n_providers"].astype(float), marker="o")
ax.set_title("Unique Providers per Year")
ax.set_xlabel("Year")
ax.set_ylabel("Providers")
ax.set_ylim(0,20000)
plt.tight_layout()
plt.show()

# Plot: total services per year (if available)
if "services" in df.columns:
    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    ax.plot(time_tbl["Year"].astype(int), time_tbl["total_services"].astype(float), marker="o")
    ax.set_title("Total Services per Year")
    ax.set_xlabel("Year")
    ax.set_ylabel("Services")
    ax.set_ylim(0,350000000)
    plt.tight_layout()
    plt.show()
````

## ✅ A2) Time coverage (2020–2023)

### What this section confirms
- **Coverage window:** The HCPCS-level dataset spans **2020 through 2023** (4 years).
- **Why 2020 is included:** 2020 is not just “extra history”. It materially supports the **lag-based architecture** (it supplies lag values for 2021 during cross-validation).

---

### Year-by-year footprint (rows, providers, services, beneficiaries)
From the year summary table:

- **Rows per year** are fairly stable but trend slightly downward:
  - 2020: **311,251** (25.72% of rows)  
  - 2021: **307,096** (25.37%)  
  - 2022: **299,919** (24.78%)  
  - 2023: **292,009** (24.13%)  

- **Unique providers per year** trend upward:
  - 2020: **18,340**
  - 2021: **18,649**
  - 2022: **18,929**
  - 2023: **19,226**

  **Interpretation:** Even though total row-count declines modestly, more distinct NPIs appear each year. That implies the dataset is becoming **more “wide” (broader provider coverage)** and slightly less “deep” per provider on average.

- **Total services per year** decline over time:
  - 2020: **~336.0M**
  - 2021: **~326.0M**
  - 2022: **~309.0M**
  - 2023: **~291.1M**

  **Interpretation:** This downward trend in total services is a **meaningful exposure shift** across years. When comparing anomalies or tail behavior across time, we should interpret raw counts in the context of this changing exposure.

A useful mental model:

- **Providers reflect breadth** (how many unique NPIs appear).

# B4) Services and beneficiaries distributions + summary stats

````python
# ============================================================
# B4) Services and beneficiaries distributions + summary stats
# ============================================================
def _quantile_table(series: pd.Series, name: str):
    s = _safe_num(series).replace([np.inf, -np.inf], np.nan).dropna()
    qs = [0.50, 0.75, 0.90, 0.95, 0.99]
    out = {f"p{int(q*100)}": float(s.quantile(q)) for q in qs} if len(s) else {}
    return pd.DataFrame([{"metric": name, **out, "n_nonnull": int(len(s))}])

qt = []
if "services" in df.columns:
    qt.append(_quantile_table(df["services"], "services"))
if "benes" in df.columns:
    qt.append(_quantile_table(df["benes"], "benes"))

if qt:
    quantiles_tbl = pd.concat(qt, ignore_index=True)
    display(quantiles_tbl)

# Histogram services (log-ish)
if "services" in df.columns:
    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    x = _safe_num(df["services"]).replace([np.inf, -np.inf], np.nan).dropna()
    ax.hist(x, bins=80)
    ax.set_title("Services distribution (raw)")
    ax.set_xlabel("services")
    ax.set_ylabel("rows")
    plt.tight_layout()
    plt.show()

    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    ax.hist(np.log1p(x), bins=80)
    ax.set_title("Services distribution (log1p)")
    ax.set_xlabel("log1p(services)")
    ax.set_ylabel("rows")
    plt.tight_layout()
    plt.show()

# Histogram benes
if "benes" in df.columns:
    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    x = _safe_num(df["benes"]).replace([np.inf, -np.inf], np.nan).dropna()
    ax.hist(x, bins=80)
    ax.set_title("Beneficiaries distribution (raw)")
    ax.set_xlabel("benes")
    ax.set_ylabel("rows")
    plt.tight_layout()
    plt.show()

    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    ax.hist(np.log1p(x), bins=80)
    ax.set_title("Beneficiaries distribution (log1p)")
    ax.set_xlabel("log1p(benes)")
    ax.set_ylabel("rows")
    plt.tight_layout()
    plt.show()
````

# B6) Slice support distribution

````python
# ============================================================
# B6) Slice support distribution
#   - Prefer existing slice_n if present
#   - Otherwise compute slice_n = peer group size for (HCPCS_Cd, Year) or our definition
# ============================================================
if "slice_n" not in df.columns:
    # Conservative default definition: HCPCS_Cd x Year peer group size
    # If our app/anomaly uses a different slice key, change grouping here.
    print("slice_n not found. Computing slice_n as count within (HCPCS_Cd, Year).")
    slice_n = (
        df.groupby(["HCPCS_Cd", "Year"], dropna=False)["Rndrng_NPI"]
          .transform("size")
    )
    df = df.assign(slice_n=_safe_num(slice_n).fillna(0).astype(int))
else:
    df["slice_n"] = _safe_num(df["slice_n"]).fillna(0).astype(int)

# Plot distribution of slice_n
fig = plt.figure(figsize=(8, 4))
ax = fig.add_subplot(111)
x = _safe_num(df["slice_n"]).replace([np.inf, -np.inf], np.nan).dropna()
ax.hist(np.log1p(x), bins=80)
ax.set_title("Slice support distribution: slice_n (log1p)")
ax.set_xlabel("log1p(slice_n)")
ax.set_ylabel("rows")
plt.tight_layout()
plt.show()

# Table: percent below thresholds
THRESHOLDS = [10, 25, 50, 100, 250, 500, 1000]
rows_total = max(int(len(df)), 1)

thr_rows = []
for t in THRESHOLDS:
    thr_rows.append({
        "threshold": f"slice_n < {t}",
        "n_rows": int((df["slice_n"] < t).sum()),
        "pct_rows": float((df["slice_n"] < t).mean() * 100),
    })

slice_support_tbl = pd.DataFrame(thr_rows)
display(slice_support_tbl)
````

## ✅ B6) Slice support distribution (how much peer support do we have per slice?)

This section quantifies **peer-group size**, which is a core reliability lever for everything downstream (tail flags, percentiles-in-slice, and “trustworthy” anomaly surfacing).

### What `slice_n` means
- A **slice** is defined as: **(HCPCS_Cd, Year)**  
- `slice_n` is the **number of row-grain records** that fall into that slice.
  - Row grain is: **(Rndrng_NPI, HCPCS_Cd, Place_Of_Srvc, Year)**
- Interpretation:
  - Larger `slice_n` means we have a **bigger peer group** to estimate percentiles and tail behavior reliably.
  - Very small `slice_n` slices have **unstable percentiles** (a single row can look “extreme” just due to small sample size).

### What we did in this notebook
- The dataset did not already contain `slice_n`, so we computed it directly from the HCPCS dataset as:
  - `slice_n = count(*) within groupby(["HCPCS_Cd", "Year"])`
- This matches the modeling-era slice definition we used later for support-aware logic.

### What the distribution shows (log1p plot)
- The histogram of `log1p(slice_n)` is **right-shifted**, meaning many slices have substantial support.
- There is still a **long left tail**, representing rare HCPCS-year combinations with limited peer support.

### Threshold table (how much data lives in small slices?)
The table below quantifies the share of rows that live in “low-support” slices:

- `slice_n < 10`: **0.72%** of rows  
- `slice_n < 25`: **1.47%** of rows  
- `slice_n < 50`: **2.82%** of rows  
- `slice_n < 100`: **4.21%** of rows  
- `slice_n < 250`: **7.09%** of rows  
- `slice_n < 500`: **11.93%** of rows  
- `slice_n < 1000`: **22.35%** of rows  

### Why this matters downstream (and how to read it)
- This is a “data defensibility” check:
  - Most rows sit in slices with **meaningful peer support**.
  - A minority of rows sit in **low-support** slices, where tail designation is less stable.
- This is exactly why our anomaly surfacing is **support-aware**:
  - We can prioritize anomalies from slices with stronger `slice_n`.
  - We can treat low-support slices more conservatively (or route them for “needs-more-context” review instead of surfacing them as high-priority).

### Practical takeaway
`slice_n` is the main guardrail against over-interpreting outliers in tiny peer groups. The distribution confirms that while low-support slices exist, they represent a relatively small fraction of total rows, and we explicitly account for them in later anomaly workflows.

````python
# ============================================================
# C–E) Cost behavior, anomaly context, and data quality (post-scoring)
# Source of truth: artifacts/eval_universe/eval_scored_DG_V3.parquet
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# 0) Load (reuse df if already in memory)
# ----------------------------
REPO_ROOT = Path.cwd()
EVAL_PATH = REPO_ROOT / "artifacts" / "eval_universe" / "eval_scored_DG_V3.parquet"

if "df" not in globals():
    df = pd.read_parquet(EVAL_PATH)
    print("[Loaded]", EVAL_PATH)
else:
    print("[Using existing df in memory]")

# Standardize expected columns where possible
# (Do not hard rename. Just validate.)
REQ_GRAIN = ["Rndrng_NPI", "HCPCS_Cd", "Place_Of_Srvc", "Year"]
for c in REQ_GRAIN:
    if c not in df.columns:
        raise ValueError(f"Missing required grain column: {c}")

df["Rndrng_NPI"] = df["Rndrng_NPI"].astype(str)
df["HCPCS_Cd"] = df["HCPCS_Cd"].astype(str)
df["Place_Of_Srvc"] = df["Place_Of_Srvc"].astype(str)
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")

# Numeric fields if present
for c in ["observed_cost", "expected_cost", "residual", "log_oe", "oe_ratio", "services", "benes", "slice_n"]:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")

# Helpful helper
def _exists(cols):
    return all(c in df.columns for c in cols)

def _show_hist(series: pd.Series, title: str, xlabel: str, log1p: bool = False, bins: int = 80):
    s = pd.to_numeric(series, errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()
    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    if log1p:
        ax.hist(np.log1p(s.clip(lower=0)), bins=bins)
        ax.set_xlabel(f"log1p({xlabel})")
    else:
        ax.hist(s, bins=bins)
        ax.set_xlabel(xlabel)
    ax.set_ylabel("rows")
    ax.set_title(title)
    plt.tight_layout()
    plt.show()
````

## ✅ C7) Observed cost distribution (HCPCS row-grain)

### What this section is measuring
At the HCPCS row grain `(Rndrng_NPI, HCPCS_Cd, Place_Of_Srvc, Year)`, `observed_cost` represents the realized (standardized) cost for that provider–code–site–year combination. Before we discuss modeling or “anomalies,” we first want to understand the **baseline distribution** of what we’re trying to benchmark.

---

### 1) Observed cost is extremely right-skewed (as expected in medical cost data)
The raw histogram shows a large mass at low costs and a long right tail (reaching very large values). This is normal for healthcare costs, where most services are inexpensive but a small fraction are very expensive.

**Implication for benchmarking:**  
- Most rows are “routine” cost levels.  
- A small tail contributes disproportionately high costs.  
- Any robust benchmarking engine must treat the tail carefully (to avoid unstable expected-cost estimates and “false alarms” driven by small support slices).

---

### 2) The log1p view shows structure that the raw view hides
The `log1p(observed_cost)` histogram is much more interpretable. It compresses extreme values and reveals the density of typical rows across the midrange.

**Implication for modeling:**  
- A log transform is useful for stability and interpretability when costs are heavy-tailed.  
- It also helps diagnostics, because we can see whether costs cluster into bands or modes that would otherwise be invisible on the raw scale.

---

### 3) Cost varies dramatically by HCPCS among high-volume codes
The “Observed cost by HCPCS (top 50 by total services)” boxplot shows that high-volume HCPCS codes can have **very different cost distributions**:
- Some codes have tight, low-cost distributions (small boxes and whiskers).
- Others show wide dispersion and/or high medians.
- A few codes appear to have much higher typical costs than the rest (clear vertical separation).

**Interpretation:**  
This is exactly why HCPCS-level modeling is necessary. Even among “common” codes, the cost behavior is not uniform. Benchmarking must compare rows to **peers within the same HCPCS context**, rather than pooling costs across codes.

---

### What to watch for (without drawing conclusions yet)
These plots are descriptive. They do **not** identify wrongdoing or intent. They simply establish:
- Costs are heavy-tailed.
- There is meaningful cross-code heterogeneity.
- The tail exists and must be modeled and evaluated explicitly.

**Next step (later sections):**  
Once expected costs are generated, we examine post-scoring distributions (`residual`, `log_oe`) and apply conservative, support-aware rules to surface rows/providers for deeper review.

# C8) Expected vs observed sanity (pre-model baseline proxy)

````python
# ============================================================
# C8) Expected vs observed sanity (pre-model baseline proxy)
#   - baseline expected-cost proxy: mean observed_cost by (HCPCS, Year)
#   - compare baseline_expected vs observed_cost distributionally
# ============================================================
if not _exists(["observed_cost", "HCPCS_Cd", "Year"]):
    print("Skipping C8: need observed_cost + HCPCS_Cd + Year.")
else:
    baseline = (
        df.dropna(subset=["observed_cost", "HCPCS_Cd", "Year"])
          .groupby(["HCPCS_Cd", "Year"], dropna=False)["observed_cost"]
          .mean()
          .rename("baseline_expected_cost")
          .reset_index()
    )

    # Merge baseline back (lightweight join)
    base_df = df.merge(baseline, on=["HCPCS_Cd", "Year"], how="left")

    # Quick sanity metrics
    tmp = base_df[["observed_cost", "baseline_expected_cost"]].copy()
    tmp = tmp.replace([np.inf, -np.inf], np.nan).dropna()
    tmp["baseline_residual"] = tmp["observed_cost"] - tmp["baseline_expected_cost"]

    display(pd.DataFrame([{
        "baseline_definition": "mean observed_cost within (HCPCS_Cd, Year)",
        "n_rows_used": int(len(tmp)),
        "baseline_residual_median": float(tmp["baseline_residual"].median()),
        "baseline_residual_p95": float(tmp["baseline_residual"].quantile(0.95)),
    }]))

    # Baseline residual histogram
    _show_hist(tmp["baseline_residual"], "Baseline proxy residual (observed - mean_by_HCPCS_Year)", "baseline_residual", log1p=False)
````

## ✅ C8) Expected vs observed sanity (pre-model baseline proxy)

### What we did (simple, transparent baseline)
Before introducing any machine learning model, we built a **naive expected-cost proxy** to sanity-check the problem:

- **Baseline expected cost** = mean `observed_cost` within each `(HCPCS_Cd, Year)` group  
- **Baseline residual** = `observed_cost - mean(observed_cost | HCPCS_Cd, Year)`

This is intentionally simple. It answers:  
> “If I only knew the code and year, how far above or below the average is each row?”

---

### Baseline summary (what our output shows)
From the table:

- **Baseline definition:** mean `observed_cost` within `(HCPCS_Cd, Year)`  
- **Rows used:** 1,210,275 (full HCPCS row-grain dataset)  
- **Median baseline residual:** ~0.0005 (very close to zero)  
- **95th percentile baseline residual:** ~20.31

**Interpretation:**
- The **median ~0** is expected because residuals are centered around the group mean by construction.
- The **p95 ~20.3** gives a simple sense of spread: within a typical `(HCPCS, Year)` slice, many rows still differ meaningfully from the slice average.

---

### How to read the histogram
The histogram of `baseline_residual` is extremely peaked around 0 with long tails.

That shape is exactly what we expect when:
- most rows are close to their slice mean, and
- a minority of rows are far above (and far below) the slice mean.

**What this tells us (and what it does not):**
- ✅ It confirms that “benchmarking within HCPCS×Year” is meaningful. There is real dispersion even after basic normalization by code and year.
- ✅ It motivates why we need a richer model (provider characteristics, geography, site of service, lag, etc.) to explain predictable variation.
- ❌ This baseline is not a production benchmark. It ignores provider, POS, state, patient mix, and temporal dynamics. It is only a sanity check.

---

### Why include this in the narrative
This baseline is a useful “control” because it is:
- **transparent** (no black-box assumptions),
- **peer-aligned** (comparisons happen within code and year),
- **a reference point** for model improvement (the ML model should reduce residual spread and improve tail behavior compared to this baseline).

**Next step:**  
We move from this naive proxy to the full scoring pipeline (D/G+V3), where expected cost is learned using richer features and explicitly evaluated in both the bulk and the tail.

# C9) Lag coverage (justifies architecture)

````python
# ============================================================
# C9) Lag coverage (justifies architecture)
#   - % rows has_lag True/False by Year
#   - cost variance / error proxy by lag bucket
# ============================================================
if "has_lag" not in df.columns:
    print("Skipping C9: has_lag not present (expected in eval_scored_DG_V3).")
else:
    # Normalize has_lag to bool-ish
    has_lag = df["has_lag"].astype(str).str.lower().isin(["true", "1", "t", "yes"])
    lag_tbl = (
        df.assign(has_lag_bool=has_lag)
          .dropna(subset=["Year"])
          .groupby(["Year", "has_lag_bool"], dropna=False)
          .size()
          .reset_index(name="n_rows")
    )
    # Shares within year
    lag_tbl["year_total"] = lag_tbl.groupby("Year")["n_rows"].transform("sum")
    lag_tbl["share_pct"] = 100 * lag_tbl["n_rows"] / lag_tbl["year_total"]
    display(lag_tbl.sort_values(["Year", "has_lag_bool"]))

    # Plot: share by year (two lines)
    pivot = lag_tbl.pivot(index="Year", columns="has_lag_bool", values="share_pct").fillna(0).sort_index()
    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    for col in pivot.columns:
        ax.plot(pivot.index.astype(int), pivot[col].astype(float), marker="o", label=f"has_lag={col}")
    ax.set_title("Lag coverage by Year (% rows with has_lag True/False)")
    ax.set_xlabel("Year")
    ax.set_ylabel("Share of rows (%)")
    ax.legend(frameon=False)
    plt.tight_layout()
    plt.show()

    # Plot: variance proxy by lag bucket (use residual if present else observed_cost)
    metric = "residual" if "residual" in df.columns else "observed_cost"
    tmp = df.copy()
    tmp["has_lag_bool"] = has_lag
    tmp[metric] = pd.to_numeric(tmp[metric], errors="coerce")

    by_lag = (
        tmp[[metric, "has_lag_bool"]]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .groupby("has_lag_bool")[metric]
        .agg(["count", "mean", "median", "std"])
        .reset_index()
        .rename(columns={"count": "n_rows"})
    )
    display(by_lag)

    fig = plt.figure(figsize=(6, 4))
    ax = fig.add_subplot(111)
    ax.bar(by_lag["has_lag_bool"].astype(str), by_lag["std"].astype(float))
    ax.set_title(f"Variance proxy by lag bucket (std of {metric})")
    ax.set_xlabel("has_lag")
    ax.set_ylabel(f"std({metric})")
    plt.tight_layout()
    plt.show()
````

## ✅ C9) Lag coverage (justifies the architecture)

### What “lag” means here
- `has_lag=True` means the row has usable historical information from a prior year at the **same row grain**:  
  **(Rndrng_NPI, HCPCS_Cd, Place_Of_Srvc, Year)**.  
- In practice, lag features let the model learn **provider-specific + code-specific persistence**, which is one of the strongest and most defensible predictors of expected cost in claims-like settings.

---

### 1) Coverage by year: how much of the dataset can use lag?

**Key result**
- **2020** has **0% lag coverage** by construction (no prior year in the dataset to reference).
- Starting in **2021**, lag coverage becomes the majority of rows and increases slightly year-over-year:
  - **2021:** `has_lag=True` for **80.3%** of rows  
  - **2022:** `has_lag=True` for **84.0%** of rows  
  - **2023:** `has_lag=True` for **85.6%** of rows  

**Interpretation**
- This validates the design choice to **include 2020** even if we don’t treat it as a primary prediction year in some evaluations.  
  It acts as the “history provider” that unlocks lag for 2021 (and onward), which materially improves model stability and realism.

---

### 2) Residual behavior differs sharply by lag bucket (a reliability signal)

From the summary table:

- **Rows with `has_lag=False`**  
  - `mean residual = 4.10`  
  - `median residual = -0.083`  
  - `std(residual) = 78.85`

- **Rows with `has_lag=True`**  
  - `mean residual = 0.092`  
  - `median residual = -0.0005`  
  - `std(residual) = 22.28`

**Interpretation (why this matters)**
- The **no-lag bucket is much noisier** (std ~79 vs ~22). That is not “bad modeling”. It is expected:
  - No-lag rows are more likely to be **cold-start** or structurally under-supported contexts (less historical anchoring).
  - Higher dispersion means expected-cost benchmarking is inherently less certain there, which is why we later lean on **support tiers, slice support, and guardrails**.
- The lag bucket is dramatically more stable and closer to unbiased (median residual ~0).  
  That’s evidence that the lag-driven architecture is not just predictive. It is also **calibration-stabilizing**, which is exactly what payers want for defensible benchmarks.

---

### Practical takeaway for the downstream system
- **Lag availability partitions the universe into two regimes**:
  1. **Lag-rich**: higher trust, tighter residual spread, more stable benchmarking.
  2. **Lag-poor**: higher variance, more conservative handling required (support tiers + guardrails + slice support rules).
- This is why we built the modeling stack to **explicitly exploit lag when present**, and why we treat no-lag regions more carefully downstream.

---

# D10) Residual and log(O/E) distributions (post-scoring)

````python
# ============================================================
# D10) Residual and log(O/E) distributions (post-scoring)
#   - hist residual, hist log_oe
#   - residual vs expected_cost scatter for a frequent HCPCS
#   - table: share positive residual, share extreme log_oe percentiles
# ============================================================
if _exists(["residual"]):
    _show_hist(df["residual"], "Residual distribution (post-scoring)", "residual", log1p=False)
else:
    print("Skipping D10 residual hist: residual not present.")

if _exists(["log_oe"]):
    _show_hist(df["log_oe"], "log(O/E) distribution (post-scoring)", "log_oe", log1p=False)
else:
    print("Skipping D10 log_oe hist: log_oe not present.")

# Scatter for frequent HCPCS (robust to object/array columns)
if _exists(["expected_cost", "residual", "HCPCS_Cd"]):
    # pick frequent HCPCS by rows (or by services if available)
    if "services" in df.columns:
        hc = df.groupby("HCPCS_Cd")["services"].sum().sort_values(ascending=False).head(1).index[0]
        hc_basis = "services"
    else:
        hc = df["HCPCS_Cd"].value_counts().idxmax()
        hc_basis = "rows"

    sub = df.loc[df["HCPCS_Cd"] == hc, ["expected_cost", "residual"]].copy()

    # Convert to numeric and drop non-finite safely (no DataFrame-wide replace)
    sub["expected_cost"] = pd.to_numeric(sub["expected_cost"], errors="coerce")
    sub["residual"] = pd.to_numeric(sub["residual"], errors="coerce")

    mask = np.isfinite(sub["expected_cost"].to_numpy()) & np.isfinite(sub["residual"].to_numpy())
    sub = sub.loc[mask]

    # cap for responsiveness
    if len(sub) > 200_000:
        sub = sub.sample(n=200_000, random_state=7)

    fig = plt.figure(figsize=(8, 4))
    ax = fig.add_subplot(111)
    ax.scatter(sub["expected_cost"], sub["residual"], s=6, alpha=0.25)
    ax.set_title(f"Residual vs expected_cost for frequent HCPCS={hc} (selected by {hc_basis})")
    ax.set_xlabel("expected_cost")
    ax.set_ylabel("residual (observed - expected)")
    plt.tight_layout()
    plt.show()
else:
    print("Skipping D10 scatter: need expected_cost + residual + HCPCS_Cd.")

# Shares table: positive residual, extreme log_oe
share_rows = {}
if "residual" in df.columns:
    r = pd.to_numeric(df["residual"], errors="coerce").replace([np.inf, -np.inf], np.nan)
    share_rows["share_positive_residual_pct"] = float((r > 0).mean() * 100)

if "log_oe" in df.columns:
    lo = pd.to_numeric(df["log_oe"], errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()
    if len(lo) > 0:
        p99 = float(lo.quantile(0.99))
        p995 = float(lo.quantile(0.995))
        share_rows["log_oe_p99"] = p99
        share_rows["log_oe_p995"] = p995
        share_rows["share_ge_p99_pct"] = float((pd.to_numeric(df["log_oe"], errors="coerce") >= p99).mean() * 100)
        share_rows["share_ge_p995_pct"] = float((pd.to_numeric(df["log_oe"], errors="coerce") >= p995).mean() * 100)

if share_rows:
    display(pd.DataFrame([share_rows]))
````

## ✅ D10) Residual and log(O/E) distributions (post-scoring)

This section moves from “what does the raw HCPCS dataset look like?” to “what does the model say is over- vs under-expected?” using the scored row-level dataset.

### What we plotted (and why it matters)

**1) Residual distribution (post-scoring)**  
- **Residual = observed_cost − expected_cost**.  
- The histogram is sharply centered near 0, which is what we want from a calibrated expected-cost model at scale.  
- The long tails represent **rows where observed and expected diverge substantially**, which is exactly the signal we later operationalize in the anomaly worklists.

**2) log(O/E) distribution (post-scoring)**  
- **O/E = observed_cost / expected_cost**, and **log(O/E)** makes multiplicative deviations easier to interpret symmetrically:
  - log(O/E) = 0 → O/E = 1 (perfectly aligned)
  - log(O/E) > 0 → over-expected
  - log(O/E) < 0 → under-expected
- The mass near 0 indicates many rows are close to parity, while the tail is where “extreme over-expected” behavior lives.

**3) Residual vs expected_cost scatter for a frequent HCPCS (J1439, selected by services)**  
- This is a “starter diagnostic” to see whether tail behavior is:
  - mostly a **multiplicative effect** (big O/E at low expected),
  - mostly an **absolute effect** (large residuals at higher expected), or
  - a mix (often the case).
- In this specific plot, the range of expected_cost is relatively tight and the residual spread is comparatively small, suggesting this HCPCS slice is fairly stable at the center, with limited but meaningful deviations.

### Key summary numbers (from the table output)

From the scored dataset:

- **Share of rows with positive residual:** **48.72%**  
  - Interpretation: the model is not trivially “biased high” or “biased low.” It produces both under- and over-expected predictions in a roughly balanced way.

- **Global log(O/E) thresholds:**
  - **p99(log_oe) = 0.4562**
  - **p99.5(log_oe) = 0.5755**

  Convert these to intuitive O/E multipliers:
  - exp(0.4562) ≈ **1.58×** (99th percentile row is ~58% over expected)
  - exp(0.5755) ≈ **1.78×** (99.5th percentile row is ~78% over expected)

- **Share of rows at or above these thresholds:**
  - **≥ p99:** **~1.00%**
  - **≥ p99.5:** **~0.50%**

  These are behaving exactly like percentile cutoffs should, which makes them a sensible “global fallback” definition of tail exposure (separate from the slice-percentile approach used in anomaly surfacing).

### 🔍 How to use this section later in the story

- This is the bridge between **benchmarking** and **anomaly surfacing**:
  - The residual/log(O/E) distributions show the **full score landscape**.
  - The extreme percentiles define a defensible “tail region.”
  - The anomaly worklists (ANOM.2 and ANOM.3) then apply **slice-aware and evidence-aware logic** to surface only the most defensible tail cases for follow-up.

**Important caution:** these plots describe statistical deviation from expected peers. They do **not** label intent or wrongdoing. Any determination of inappropriate billing requires structured clinical and coding review, documentation checks, and payer policy context.

# D11) Tail exposure check

````python
# ============================================================
# D11) Tail exposure check
#   - counts above robust threshold by Year
#   - tail share of spend proxy (sum observed_cost for tail vs overall)
#   - uses log_oe_pct_in_slice (if present) or falls back to global log_oe p99
# ============================================================

if not _exists(["observed_cost", "Year"]):
    print("Skipping D11: need observed_cost + Year.")
else:
    # Only pull columns we need (avoids array/object columns breaking replace)
    need_cols = ["Year", "observed_cost"]
    if "log_oe_pct_in_slice" in df.columns:
        need_cols.append("log_oe_pct_in_slice")
    if "log_oe" in df.columns:
        need_cols.append("log_oe")

    tmp = df[need_cols].copy()

    tmp["observed_cost"] = pd.to_numeric(tmp["observed_cost"], errors="coerce")
    tmp["Year"] = pd.to_numeric(tmp["Year"], errors="coerce")

    # Drop non-finite safely
    mask = np.isfinite(tmp["observed_cost"].to_numpy()) & np.isfinite(tmp["Year"].to_numpy())
    tmp = tmp.loc[mask].copy()
    tmp["Year"] = tmp["Year"].astype(int)

    # Prefer slice percentile if available (aligned with robust logic)
    if "log_oe_pct_in_slice" in tmp.columns:
        tmp["log_oe_pct_in_slice"] = pd.to_numeric(tmp["log_oe_pct_in_slice"], errors="coerce")
        tmp["is_tail_robust_proxy"] = tmp["log_oe_pct_in_slice"] >= 0.99
        tail_def = "log_oe_pct_in_slice >= 0.99 (proxy for top 1% in slice)"
    elif "log_oe" in tmp.columns:
        tmp["log_oe"] = pd.to_numeric(tmp["log_oe"], errors="coerce")
        lo = tmp["log_oe"].replace([np.inf, -np.inf], np.nan).dropna()
        thr = float(lo.quantile(0.99)) if len(lo) else np.nan
        tmp["is_tail_robust_proxy"] = tmp["log_oe"] >= thr
        tail_def = f"global log_oe >= p99 (fallback), threshold={thr:.4f}" if np.isfinite(thr) else "global log_oe >= p99 (fallback)"
    else:
        print("Skipping D11: need log_oe_pct_in_slice or log_oe.")
        tmp = None

    if tmp is not None:
        tail_by_year = (
            tmp.groupby("Year")["is_tail_robust_proxy"]
               .agg(n_tail="sum", n_rows="count")
               .reset_index()
               .sort_values("Year")
        )
        tail_by_year["tail_share_rows_pct"] = 100 * tail_by_year["n_tail"] / tail_by_year["n_rows"]

        display(pd.DataFrame([{"tail_definition": tail_def}]))
        display(tail_by_year)

        fig = plt.figure(figsize=(8, 4))
        ax = fig.add_subplot(111)
        ax.plot(tail_by_year["Year"].astype(int), tail_by_year["n_tail"].astype(float), marker="o")
        ax.set_title("Tail row counts by Year (robust proxy)")
        ax.set_xlabel("Year")
        ax.set_ylabel("Tail rows")
        plt.tight_layout()
        plt.show()

        # Spend proxy: sum observed_cost in tail vs overall
        spend_tbl = (
            tmp.assign(is_tail=tmp["is_tail_robust_proxy"].fillna(False))
               .groupby("is_tail")["observed_cost"]
               .sum()
               .reset_index()
               .rename(columns={"observed_cost": "sum_observed_cost"})
        )
        total = float(spend_tbl["sum_observed_cost"].sum())
        spend_tbl["share_of_total_pct"] = 100 * spend_tbl["sum_observed_cost"] / max(total, 1.0)
        display(spend_tbl)
````

**Interpretation**
- For the fields that matter most to expected-cost benchmarking and anomaly analysis (costs, residuals, log(O/E), volume, and key identifiers), **missingness is effectively zero**.
- This is a strong reliability signal. It means downstream computations like `residual = observed - expected`, `oe_ratio`, `log_oe`, slice-support, and provider rollups are not being driven by imputation or row drops.

**Why this matters**
- Missingness is one of the fastest ways to create accidental “anomalies” (for example, missing expected_cost leading to undefined O/E, or missing services leading to unstable support tiers). That risk is minimal here.

---

### ⚠️ Invalid or problematic expected_cost values

| expected_cost <= 0 count | expected_cost <= 0 pct | expected_cost isna pct | note |
|---:|---:|---:|---|
| 39 | 0.003222% | 0.0% | log_oe is only meaningful when expected_cost > 0 ... |

**Interpretation**
- There are **39 rows** where `expected_cost <= 0` (about **0.0032%** of rows).
- This is tiny in prevalence, but it is important conceptually because:
  - `oe_ratio = observed_cost / expected_cost` is undefined or sign-flipped when `expected_cost <= 0`
  - `log_oe = log(oe_ratio)` is only meaningful when `expected_cost > 0` and `oe_ratio > 0`

**How we handle it (principle)**
- We treat these as **edge-case numerical validity issues**, not signal.
- The defensible approach is:
  - keep `expected_cost` as-is for auditability,
  - but **avoid interpreting log(O/E)** for these rows (and ensure any log-based logic either excludes them or sets log_oe to NaN for those cases).

**Practical implication**
- This does *not* impact the dataset in a material way (extremely small share), but it is a good example of why we built:
  - explicit validity checks,
  - guardrails and conservative anomaly definitions,
  - and consistent handling rules in scoring and surfacing.

---

### Takeaway
- **Missingness is not a concern** in this HCPCS-level dataset for the key fields we rely on.
- The only data-quality flag of note is a **very small number of `expected_cost <= 0` rows**, which we treat as non-interpretable for log-based ratios and handle defensibly in downstream logic.

# E13) Consistency across years

````python
# ============================================================
# E13) Consistency across years
#   - median observed_cost and expected_cost by Year for top HCPCS
# ============================================================

if _exists(["observed_cost", "expected_cost", "HCPCS_Cd", "Year"]):
    # top HCPCS by services if available else by rows
    if "services" in df.columns:
        top_h = (
            df.groupby("HCPCS_Cd")["services"].sum().sort_values(ascending=False).head(10).index.tolist()
        )
        top_basis = "services"
    else:
        top_h = df["HCPCS_Cd"].value_counts().head(10).index.tolist()
        top_basis = "rows"

    # Only keep needed columns (prevents replace() crashing on array/object cols)
    cols = ["HCPCS_Cd", "Year", "observed_cost", "expected_cost"]
    sub = df.loc[df["HCPCS_Cd"].isin(top_h), cols].copy()

    # Coerce numerics and drop non-finite cleanly
    sub["Year"] = pd.to_numeric(sub["Year"], errors="coerce")
    sub["observed_cost"] = pd.to_numeric(sub["observed_cost"], errors="coerce")
    sub["expected_cost"] = pd.to_numeric(sub["expected_cost"], errors="coerce")

    mask = (
        np.isfinite(sub["Year"].to_numpy())
        & np.isfinite(sub["observed_cost"].to_numpy())
        & np.isfinite(sub["expected_cost"].to_numpy())
    )
    sub = sub.loc[mask].copy()
    sub["Year"] = sub["Year"].astype(int)

    yr_tbl = (
        sub.groupby(["HCPCS_Cd", "Year"], dropna=False)
           .agg(
               median_observed_cost=("observed_cost", "median"),
               median_expected_cost=("expected_cost", "median"),
               n_rows=("HCPCS_Cd", "size"),
           )
           .reset_index()
           .sort_values(["HCPCS_Cd", "Year"])
    )

    display(pd.DataFrame([{"top_hcpcs_selected_by": top_basis, "top_hcpcs": top_h}]))
    display(yr_tbl.head(30))

    # Plot: for each top HCPCS, median observed vs expected across years
    for hc in top_h:
        t = yr_tbl[yr_tbl["HCPCS_Cd"] == hc].copy()
        if len(t) == 0:
            continue
        fig = plt.figure(figsize=(8, 4))
        ax = fig.add_subplot(111)
        ax.plot(t["Year"].astype(int), t["median_observed_cost"].astype(float), marker="o", label="median observed_cost")
        ax.plot(t["Year"].astype(int), t["median_expected_cost"].astype(float), marker="o", label="median expected_cost")
        ax.set_title(f"Year consistency: HCPCS={hc} (median costs)")
        ax.set_xlabel("Year")
        ax.set_ylabel("Cost")
        ax.legend(frameon=False)
        plt.tight_layout()
        plt.show()
else:
    print("Skipping E13: need observed_cost + expected_cost + HCPCS_Cd + Year.")
````

## ✅ E13) Consistency across years (top HCPCS by service volume)

### What this section checks
This is a **stability sanity check**: for the highest-volume HCPCS codes (selected by total services), we compare **median observed_cost vs median expected_cost** over time.

Goal: confirm the model’s expected-cost signal is **directionally aligned** with observed costs across years, while also surfacing any **year-to-year shifts** that may reflect policy changes, pricing drift, coding practice changes, or composition changes (providers, sites of service).

---

### Inputs and how “top HCPCS” were chosen
- **Selection rule:** `top_hcpcs_selected_by = services`
- Interpretation: these are the codes with the largest exposure, so they are the most important for payer-facing benchmarking (they drive a lot of volume and spend opportunity).

---

### What the output table represents
Each row is:
- one HCPCS code
- one year
- median observed cost in that HCPCS-year
- median expected cost in that HCPCS-year
- number of row-grain observations supporting that median

**Columns**
- `HCPCS_Cd`: procedure/drug code
- `Year`: 2020–2023
- `median_observed_cost`: median of observed allowed amount (row-grain)
- `median_expected_cost`: median model-predicted expected cost (row-grain)
- `n_rows`: number of rows behind that HCPCS-year median (reliability proxy)

---

### How to read these results (what “good” looks like)
For a given HCPCS:
- If **median_expected_cost tracks median_observed_cost** over years, it suggests the model learned stable structure and is not drifting arbitrarily.
- If expected is consistently above or below observed, that can be:
  - a systematic bias for that code,
  - a composition issue (changing provider mix, POS mix),
  - or a real shift not captured by features.
- **n_rows matters**. Higher counts mean the medians are more stable and trustworthy.

---

### What we see in the provided outputs (examples)
Below are interpretations strictly based on the table above.

#### ✅ Strong tracking patterns (expected tracks observed)
- **J9271** (high and increasing): observed and expected both rise from 2020 → 2023 and stay close each year.
  - This is the ideal “stability” signature: the model captures year trend and level.
- **J0897** (mid/high and increasing): observed and expected move upward together across all years and remain close.
- **J1439** (stable level): observed and expected are tightly aligned each year, with minimal drift.

#### ⚠️ Noticeable year-to-year regime shift or step-change
- **J1453**: large drop from 2020 → 2021 in observed cost (≈1.00 → ≈0.26) with expected also dropping sharply.
  - Interpretation: likely a **structural shift** (coding/payment rule, pricing, or composition change). The model adjusts directionally, which is good. The gap in 2020 (observed notably higher than expected) suggests possible baseline mismatch for that year.
- **J1756**: expected is much higher than observed in 2020, then converges closely from 2021 onward.
  - Interpretation: 2020 may be atypical for this code (or lag/feature availability differs). After 2020 the model stabilizes.

#### ✅ Mostly aligned with small deviations
- **J0881**: expected and observed are generally close, with mild divergence in 2021 (expected > observed), then reconverging by 2023.
- **J9267**: expected starts higher than observed in 2020, then becomes very close by 2023.

---

### Why this matters for the benchmarking engine
- This supports the claim that **expected cost is not just a black box output**, it is **stable and interpretable over time** for high-exposure codes.
- It also highlights where year-specific behavior may warrant:
  - closer feature review (e.g., POS/state/provider-mix shifts),
  - guardrail reliance (especially for cold/low-lag conditions),
  - or segment-specific modeling improvements.

### E13) Consistency across years (First 4 plots)

These plots compare **median observed_cost** vs **median expected_cost** over time for high-volume HCPCS codes (selected by total services). The goal is not to force perfect overlap. It is to sanity-check that the model’s *central tendency* tracks the observed costs reasonably across years, while still leaving room for residual signal that downstream anomaly surfacing can detect.

#### HCPCS = J1439 (median costs)
- **What the plot shows:** Median observed_cost is fairly stable from 2020–2023, while median expected_cost jumps up from 2020 to 2021 and then closely tracks observed thereafter.
- **How to interpret:**
  - 2020 looks like a **cold-start / weaker-lag** year effect for this code. The expected curve is lower in 2020, then “snaps” into alignment once lag becomes available in later years.
  - Post-2021, the model’s expected_cost is **well-calibrated at the median** for this HCPCS, which supports using residual/log(O/E) as the anomaly signal at the row level.
- **Why this matters:** This is the architecture doing what we designed it to do. With lag in play, the expected curve becomes more stable and accurate.

#### HCPCS = Q0138 (median costs)
- **What the plot shows:** Both observed and expected medians **decline year-over-year**, with the two lines staying very close (same direction, similar magnitude).
- **How to interpret:**
  - The model is capturing **a true drift over time** (whatever the underlying driver is. pricing, case mix, coding dynamics, or policy changes).
  - The tight tracking suggests good **temporal generalization** for this code. It is not simply memorizing one year’s level.
- **Why this matters:** If the model failed to adapt, we would see persistent over- or under-shooting in later years. Here we see a coherent time trend with strong agreement.

#### HCPCS = J9271 (median costs)
- **What the plot shows:** Both observed and expected medians **increase steadily from 2020 to 2023**, and they track each other closely (expected slightly below observed but very consistent).
- **How to interpret:**
  - This looks like a clean example of the model learning a **stable mapping** and preserving it across years, even as the overall level changes.
  - The small, consistent gap (observed above expected) can be interpreted as **slight underprediction at the median**, which is not necessarily bad. What matters more is stability and not generating artifacts (e.g., flipping signs or drifting wildly).
- **Why this matters:** This is a high-volume code where the model’s expected baseline needs to be dependable. Tight tracking here increases trust in residual-based ranking and percentile-in-slice logic later.

#### HCPCS = J0881 (median costs)
- **What the plot shows:** Both observed and expected medians **trend downward** across years (with some divergence in 2021–2022, then reconverging by 2023).
- **How to interpret:**
  - The model follows the direction of change, which is the first requirement for temporal stability.
  - The mid-period separation suggests either (i) a code-year dynamic not fully captured by features, (ii) a shift in the HCPCS slice composition, or (iii) lag coverage changes influencing fit quality.
- **Why this matters:** This is exactly the kind of HCPCS where we want downstream anomaly surfacing to be conservative and slice-size aware, because the baseline is stable enough to use, but not perfect.

---

**Takeaway across these examples:** The model’s expected_cost generally tracks observed_cost medians across years for high-volume codes, with the largest discrepancies tending to appear in earlier years (consistent with weaker lag availability and cold-start dynamics). This supports using **row-level residual and log(O/E)** as the downstream signal, because the baseline behaves like a sensible expected-cost anchor rather than a noisy proxy.

````python
# ============================================================
# Sanity check: DG+V3 (final expected_cost) vs Baselines A–D
# (PATCHED to avoid pandas .replace([np.inf, -np.inf], np.nan) ambiguity)
# ============================================================

from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error
import numpy as np
import pandas as pd

# ----------------------------
# 0) Load scored universe
# ----------------------------
EVAL_PATH = "artifacts/eval_universe/eval_scored_DG_V3.parquet"
df0 = pd.read_parquet(EVAL_PATH)

# ----------------------------
# 1) Column mapping (adjust here only if needed)
# ----------------------------
group_col = "Rndrng_NPI"
target_col = "observed_cost"      # y_true
dg_pred_col = "expected_cost"     # DG+V3 prediction
lag_col = "lag1_avg_amt"          # lag-only baseline (may or may not exist)
provider_type_col = "provider_type"
hcpcs_col = "HCPCS_Cd"
pos_col = "Place_Of_Srvc"
state_col = "state"
year_col = "Year"

# Optional explicit tail flag (if present). If not present, we compute it from TRAIN.
explicit_tail_flag_candidates = [
    "is_top_1pct_avg_mdcr_stdzd_amt",
    "is_top_1pct_stdzd_amt_per_service",
    "is_top_1pct_observed_cost",
]
tail_flag_col = next((c for c in explicit_tail_flag_candidates if c in df0.columns), None)

# ----------------------------
# 2) Basic validation + cleaning
# ----------------------------
required = [group_col, target_col, dg_pred_col, provider_type_col, hcpcs_col, pos_col, state_col, year_col]
missing = [c for c in required if c not in df0.columns]
if missing:
    raise ValueError(f"Missing required columns in eval_scored parquet: {missing}")

df = df0.copy()

# Coerce numerics we care about
df[target_col] = pd.to_numeric(df[target_col], errors="coerce")
df[dg_pred_col] = pd.to_numeric(df[dg_pred_col], errors="coerce")
if lag_col in df.columns:
    df[lag_col] = pd.to_numeric(df[lag_col], errors="coerce")

# IMPORTANT FIX:
# Only replace inf/-inf in numeric columns we will use.
num_cols_to_clean = [target_col, dg_pred_col]
if lag_col in df.columns:
    num_cols_to_clean.append(lag_col)

df[num_cols_to_clean] = df[num_cols_to_clean].replace([np.inf, -np.inf], np.nan)

# Drop rows where required keys or y_true are missing
df = df.dropna(subset=[target_col, group_col, provider_type_col, hcpcs_col, pos_col, state_col, year_col]).copy()

# Standardize key dtypes
df[group_col] = df[group_col].astype(str)
df[provider_type_col] = df[provider_type_col].astype(str)
df[hcpcs_col] = df[hcpcs_col].astype(str)
df[pos_col] = df[pos_col].astype(str)
df[state_col] = df[state_col].astype(str)
df[year_col] = df[year_col].astype(int)

# ----------------------------
# 3) New-provider split (GroupShuffleSplit on NPI)
# ----------------------------
test_size = 0.20
random_state = 0
w_tail = 10

groups_all = df[group_col].to_numpy()
gss = GroupShuffleSplit(n_splits=1, test_size=test_size, random_state=random_state)
train_idx, test_idx = next(gss.split(df, groups=groups_all))

train_df = df.iloc[train_idx].copy()
test_df = df.iloc[test_idx].copy()

# ----------------------------
# 4) Tail flag logic (explicit or computed from TRAIN)
# ----------------------------
def _attach_tail_flag(train_df: pd.DataFrame, test_df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame, str]:
    if tail_flag_col is not None:
        train_df2 = train_df.copy()
        test_df2 = test_df.copy()
        train_df2["_is_tail"] = train_df2[tail_flag_col].astype(bool)
        test_df2["_is_tail"] = test_df2[tail_flag_col].astype(bool)
        return train_df2, test_df2, f"explicit tail flag: {tail_flag_col}"
    else:
        thr = float(train_df[target_col].quantile(0.99))
        train_df2 = train_df.copy()
        test_df2 = test_df.copy()
        train_df2["_is_tail"] = train_df2[target_col] >= thr
        test_df2["_is_tail"] = test_df2[target_col] >= thr
        return train_df2, test_df2, f"computed tail flag: y_true >= train p99 (thr={thr:.6g})"

train_df, test_df, tail_desc = _attach_tail_flag(train_df, test_df)

# ----------------------------
# 5) Shared metric function (mirrors modeling notebook)
# ----------------------------
def _eval_metrics(y_true: pd.Series, pred: pd.Series, is_tail: pd.Series, w_tail: int) -> dict:
    y_true = pd.to_numeric(y_true, errors="coerce")
    pred = pd.to_numeric(pred, errors="coerce")
    is_tail = is_tail.astype(int)

    m = {}
    m["r2_test"] = r2_score(y_true, pred)
    m["mae"] = mean_absolute_error(y_true, pred)
    m["rmse"] = root_mean_squared_error(y_true, pred)

    pred_clip = pred.clip(lower=0)
    m["r2_test_log"] = r2_score(np.log1p(y_true), np.log1p(pred_clip))
    m["mae_log"] = mean_absolute_error(np.log1p(y_true), np.log1p(pred_clip))
    m["rmse_log"] = root_mean_squared_error(np.log1p(y_true), np.log1p(pred_clip))

    sample_w = (1 + (w_tail - 1) * is_tail).astype(float)
    m["r2_test_w"] = r2_score(y_true, pred, sample_weight=sample_w)
    m["mae_w"] = mean_absolute_error(y_true, pred, sample_weight=sample_w)
    m["rmse_w"] = root_mean_squared_error(y_true, pred, sample_weight=sample_w)

    df_eval = pd.DataFrame({"y": y_true, "pred": pred, "is_tail": is_tail})
    df_eval["sq_err"] = (df_eval["y"] - df_eval["pred"]) ** 2

    non_tail = df_eval[df_eval["is_tail"] == 0]
    tail = df_eval[df_eval["is_tail"] == 1]

    def _safe_mean(s):
        s = pd.to_numeric(s, errors="coerce")
        return float(s.mean()) if len(s) else np.nan

    m["non_tail_pred_to_y_ratio"] = _safe_mean(non_tail["pred"]) / max(_safe_mean(non_tail["y"]), 1e-12)
    m["tail_pred_to_y_ratio"] = _safe_mean(tail["pred"]) / max(_safe_mean(tail["y"]), 1e-12)

    non_tail_sse = float(non_tail["sq_err"].sum())
    tail_sse = float(tail["sq_err"].sum())
    tot_sse = non_tail_sse + tail_sse

    m["non_tail_sse"] = non_tail_sse
    m["tail_sse"] = tail_sse
    m["non_tail_share_of_total_sse"] = non_tail_sse / max(tot_sse, 1e-12)
    m["tail_share_of_total_sse"] = tail_sse / max(tot_sse, 1e-12)

    m["non_tail_under_prediction_rate"] = float((non_tail["pred"] < non_tail["y"]).mean()) if len(non_tail) else np.nan
    m["tail_under_prediction_rate"] = float((tail["pred"] < tail["y"]).mean()) if len(tail) else np.nan

    m["non_tail_bias"] = _safe_mean(non_tail["pred"]) - _safe_mean(non_tail["y"])
    m["tail_bias"] = _safe_mean(tail["pred"]) - _safe_mean(tail["y"])

    return m

# ----------------------------
# 6) Baseline predictions (train-derived lookups)
# ----------------------------
group_cols_B = [provider_type_col, hcpcs_col, pos_col]
pt_pos_state_cols_C = [provider_type_col, pos_col, state_col]

group_cols_D = [provider_type_col, hcpcs_col, pos_col, state_col]
grp_no_state_cols_D = [provider_type_col, hcpcs_col, pos_col]
hcpcs_cols_D = [hcpcs_col]
pt_pos_cols_D = [provider_type_col, pos_col]
pt_cols_D = [provider_type_col]

def _baseline_A_pred(test_df: pd.DataFrame) -> pd.Series:
    if lag_col not in test_df.columns:
        return pd.Series(index=test_df.index, dtype=float)
    return pd.to_numeric(test_df[lag_col], errors="coerce")

def _baseline_B_pred(train_df: pd.DataFrame, test_df: pd.DataFrame) -> pd.Series:
    train_means_grp = train_df.groupby(group_cols_B)[target_col].mean().rename("mean_train_grp").reset_index()
    train_means_hcpcs = train_df.groupby([hcpcs_col])[target_col].mean().rename("mean_train_hcpcs").reset_index()
    global_mean = float(train_df[target_col].mean())

    t = test_df.copy()
    t["_orig_idx"] = t.index
    t = t.merge(train_means_grp, on=group_cols_B, how="left")
    t = t.merge(train_means_hcpcs, on=hcpcs_col, how="left")
    t = t.set_index("_orig_idx", drop=True).loc[test_df.index]

    pred = pd.to_numeric(t.get(lag_col, np.nan), errors="coerce")
    pred = pred.fillna(t["mean_train_grp"])
    pred = pred.fillna(t["mean_train_hcpcs"])
    pred = pred.fillna(global_mean)
    return pred.astype(float)

def _baseline_C_pred(train_df: pd.DataFrame, test_df: pd.DataFrame) -> pd.Series:
    train_means = (
        train_df.groupby(pt_pos_state_cols_C)[target_col]
        .mean()
        .rename("mean_train_pt_pos_state")
        .reset_index()
    )
    global_mean = float(train_df[target_col].mean())

    t = test_df.copy()
    t["_orig_idx"] = t.index
    t = t.merge(train_means, on=pt_pos_state_cols_C, how="left")
    t = t.set_index("_orig_idx", drop=True).loc[test_df.index]

    return t["mean_train_pt_pos_state"].fillna(global_mean).astype(float)

def _baseline_D_pred(train_df: pd.DataFrame, test_df: pd.DataFrame) -> pd.Series:
    mean_grp_state = train_df.groupby(group_cols_D)[target_col].mean().rename("mean_grp_state").reset_index()
    mean_grp = train_df.groupby(grp_no_state_cols_D)[target_col].mean().rename("mean_grp").reset_index()
    mean_hcpcs = train_df.groupby(hcpcs_cols_D)[target_col].mean().rename("mean_hcpcs").reset_index()
    mean_pt_pos_state = train_df.groupby(pt_pos_state_cols_C)[target_col].mean().rename("mean_pt_pos_state").reset_index()
    mean_pt_pos = train_df.groupby(pt_pos_cols_D)[target_col].mean().rename("mean_pt_pos").reset_index()
    mean_pt = train_df.groupby(pt_cols_D)[target_col].mean().rename("mean_pt").reset_index()
    global_mean = float(train_df[target_col].mean())

    t = test_df.copy()
    t["_orig_idx"] = t.index
    t = t.merge(mean_grp_state, on=group_cols_D, how="left")
    t = t.merge(mean_grp, on=grp_no_state_cols_D, how="left")
    t = t.merge(mean_hcpcs, on=hcpcs_cols_D, how="left")
    t = t.merge(mean_pt_pos_state, on=pt_pos_state_cols_C, how="left")
    t = t.merge(mean_pt_pos, on=pt_pos_cols_D, how="left")
    t = t.merge(mean_pt, on=pt_cols_D, how="left")
    t = t.set_index("_orig_idx", drop=True).loc[test_df.index]

    pred = t["mean_grp_state"]
    pred = pred.fillna(t["mean_grp"])
    pred = pred.fillna(t["mean_hcpcs"])
    pred = pred.fillna(t["mean_pt_pos_state"])
    pred = pred.fillna(t["mean_pt_pos"])
    pred = pred.fillna(t["mean_pt"])
    pred = pred.fillna(global_mean).astype(float)
    return pred

def _dg_v3_pred(test_df: pd.DataFrame) -> pd.Series:
    return pd.to_numeric(test_df[dg_pred_col], errors="coerce")

# ----------------------------
# 7) Run evaluation on TEST and summarize
# ----------------------------
y_true = test_df[target_col].astype(float)
is_tail_test = test_df["_is_tail"].astype(int)

preds = {}

# Baseline A: evaluate only rows with lag present, mirroring our Baseline A notebook cell
pred_A = _baseline_A_pred(test_df)
mask_A = pred_A.notna() & (pred_A >= 0)
preds["Baseline_A_lag_only"] = (pred_A[mask_A], y_true[mask_A], is_tail_test[mask_A])

# Baseline B/C/D: evaluate all rows (they backoff)
preds["Baseline_B_lag_plus_backoff"] = (_baseline_B_pred(train_df, test_df), y_true, is_tail_test)
preds["Baseline_C_pt_pos_state_mean"] = (_baseline_C_pred(train_df, test_df), y_true, is_tail_test)
preds["Baseline_D_full_backoff_ladder"] = (_baseline_D_pred(train_df, test_df), y_true, is_tail_test)

# DG+V3: evaluate where expected_cost exists (should be ~all)
pred_DG = _dg_v3_pred(test_df)
mask_DG = pred_DG.notna()
preds["DG_plus_V3_expected_cost"] = (pred_DG[mask_DG], y_true[mask_DG], is_tail_test[mask_DG])

rows = []
for name, (p, y, tflag) in preds.items():
    m = _eval_metrics(y.astype(float), p.astype(float), tflag.astype(int), w_tail=w_tail)
    m["model"] = name
    m["n_test_rows_used"] = int(len(y))
    rows.append(m)

summary = pd.DataFrame(rows).set_index("model").sort_values("rmse_w")

display(pd.DataFrame([{
    "tail_definition_used": tail_desc,
    "test_size": test_size,
    "random_state": random_state,
    "w_tail": w_tail,
    "n_train_rows": int(len(train_df)),
    "n_test_rows_total": int(len(test_df)),
    "n_test_rows_DG_used": int(mask_DG.sum()),
    "n_test_rows_baseline_A_used": int(mask_A.sum()),
}]))

key_cols = [
    "n_test_rows_used",
    "r2_test", "mae", "rmse",
    "r2_test_log", "mae_log", "rmse_log",
    "r2_test_w", "mae_w", "rmse_w",
    "tail_share_of_total_sse",
    "tail_under_prediction_rate",
    "tail_pred_to_y_ratio",
    "tail_bias",
]
display(summary[key_cols])

# ----------------------------
# 8) “Proof-style” comparisons (prints whether DG wins on tail-sensitive metrics)
# ----------------------------
dg = summary.loc["DG_plus_V3_expected_cost"]

def _compare(metric: str, lower_is_better: bool):
    out = []
    for mname in summary.index:
        if mname == "DG_plus_V3_expected_cost":
            continue
        other = summary.loc[mname]
        if lower_is_better:
            win = float(dg[metric]) < float(other[metric])
        else:
            win = float(dg[metric]) > float(other[metric])
        out.append({
            "metric": metric,
            "baseline": mname,
            "DG_wins": bool(win),
            "DG": float(dg[metric]),
            "baseline_value": float(other[metric]),
        })
    return pd.DataFrame(out)

display(_compare("rmse_w", lower_is_better=True).sort_values(["DG_wins", "baseline"], ascending=[False, True]))
display(_compare("tail_share_of_total_sse", lower_is_better=True).sort_values(["DG_wins", "baseline"], ascending=[False, True]))
display(_compare("tail_under_prediction_rate", lower_is_better=True).sort_values(["DG_wins", "baseline"], ascending=[False, True]))

print("\nInterpretation tip:")
print("- Prioritize tail-weighted error (rmse_w/mae_w), tail SSE share, and tail underprediction rate.")
print("- If DG+V3 improves those vs A–D on the same split, that supports the selection rationale.")

# ============================================================
# ADD-ON (apples-to-apples): Compare DG/B/C/D on the SAME lag-available subset as Baseline A
# Goal:
#   - Baseline A (lag-only) is evaluated only where lag exists (mask_A).
#   - To make the comparison fair, evaluate DG, B, C, D on the exact same subset.
#   - Also report coverage: what share of the test set has valid lag.
# Where to place:
#   - Put this block immediately AFTER our existing sanity-check cell,
#     right after we compute mask_A and display the main summary.
# ============================================================

# --- Coverage of lag in the test set (how much of the world Baseline A can even serve) ---
lag_coverage = float(mask_A.mean() * 100)
display(pd.DataFrame([{
    "note": "Lag coverage in test set (Baseline A can only score on this subset)",
    "n_test_rows_total": int(len(test_df)),
    "n_test_rows_with_valid_lag": int(mask_A.sum()),
    "lag_coverage_pct": lag_coverage,
}]))

# --- Re-evaluate DG/B/C/D on the same lag-available subset ---
y_A = y_true[mask_A]
t_A = is_tail_test[mask_A]

# Use the already-built baseline predictions on the full test set, then subset
pred_B_full = preds["Baseline_B_lag_plus_backoff"][0]
pred_C_full = preds["Baseline_C_pt_pos_state_mean"][0]
pred_D_full = preds["Baseline_D_full_backoff_ladder"][0]
pred_DG_full = preds["DG_plus_V3_expected_cost"][0]

# IMPORTANT:
# pred_DG_full was built on mask_DG (expected_cost notna). Ensure same index alignment:
pred_DG_full = pred_DG_full.reindex(test_df.index)

# Subset all models to the lag-valid subset
preds_lag_subset = {
    "Baseline_A_lag_only": pred_A[mask_A],
    "Baseline_B_lag_plus_backoff": pred_B_full[mask_A],
    "Baseline_C_pt_pos_state_mean": pred_C_full[mask_A],
    "Baseline_D_full_backoff_ladder": pred_D_full[mask_A],
    "DG_plus_V3_expected_cost": pred_DG_full[mask_A],
}

rows_subset = []
for name, p in preds_lag_subset.items():
    # Drop any NA preds just in case (should be rare for B/C/D/DG, common pattern safety)
    ok = p.notna()
    m = _eval_metrics(
        y_A[ok].astype(float),
        p[ok].astype(float),
        t_A[ok].astype(int),
        w_tail=w_tail,
    )
    m["model"] = name
    m["n_test_rows_used"] = int(ok.sum())
    rows_subset.append(m)

summary_subset = (
    pd.DataFrame(rows_subset)
    .set_index("model")
    .sort_values("rmse_w")
)

display(pd.DataFrame([{
    "subset": "Lag-available test rows only (mask_A)",
    "n_rows_subset": int(mask_A.sum()),
    "tail_definition_used": tail_desc,
    "w_tail": w_tail,
}]))

key_cols_subset = [
    "n_test_rows_used",
    "r2_test", "mae", "rmse",
    "r2_test_log", "mae_log", "rmse_log",
    "r2_test_w", "mae_w", "rmse_w",
    "tail_share_of_total_sse",
    "tail_under_prediction_rate",
    "tail_pred_to_y_ratio",
    "tail_bias",
]
display(summary_subset[key_cols_subset])

# --- Optional: direct “DG vs Baseline A” deltas on the lag subset (easy narrative) ---
if "DG_plus_V3_expected_cost" in summary_subset.index and "Baseline_A_lag_only" in summary_subset.index:
    dg_s = summary_subset.loc["DG_plus_V3_expected_cost"]
    a_s = summary_subset.loc["Baseline_A_lag_only"]

    deltas = pd.DataFrame([{
        "metric": "rmse_w (lower better)",
        "DG": float(dg_s["rmse_w"]),
        "Baseline_A": float(a_s["rmse_w"]),
        "DG_minus_A": float(dg_s["rmse_w"] - a_s["rmse_w"]),
    },{
        "metric": "tail_under_prediction_rate (lower better)",
        "DG": float(dg_s["tail_under_prediction_rate"]),
        "Baseline_A": float(a_s["tail_under_prediction_rate"]),
        "DG_minus_A": float(dg_s["tail_under_prediction_rate"] - a_s["tail_under_prediction_rate"]),
    },{
        "metric": "tail_pred_to_y_ratio (closer to 1 is better)",
        "DG": float(dg_s["tail_pred_to_y_ratio"]),
        "Baseline_A": float(a_s["tail_pred_to_y_ratio"]),
        "DG_minus_A": float(dg_s["tail_pred_to_y_ratio"] - a_s["tail_pred_to_y_ratio"]),
    },{
        "metric": "tail_bias (closer to 0 is better)",
        "DG": float(dg_s["tail_bias"]),
        "Baseline_A": float(a_s["tail_bias"]),
        "DG_minus_A": float(dg_s["tail_bias"] - a_s["tail_bias"]),
    }])

    display(deltas)

print(
    "\nInterpretation guide:\n"
    "1) Baseline A usually wins on the lag-only subset because lag is extremely predictive.\n"
    "2) The point of DG+V3 is that it stays competitive on lag rows AND works for 100% of rows (cold-start included).\n"
    "3) Use the lag_coverage_pct above to highlight why a pure lag-only baseline is not a complete production solution."
)
````

## 🏁 Interpretation: DG+V3 vs Baselines A–D (full output + lag-subset add-on)

### 0) What this sanity check is actually testing
- **Data**: `eval_scored_DG_V3.parquet` (post-scoring universe at row grain).
- **Test setting**: **new-provider split** (`GroupShuffleSplit` on `Rndrng_NPI`, test_size=0.20).  
  This is intentionally harder and more realistic for deployment because it measures generalization to new providers.
- **Target**: `observed_cost`.
- **DG+V3 prediction**: `expected_cost` from the final hybrid architecture + guardrails.
- **Tail emphasis**: `w_tail = 10` means tail rows are weighted 10x in the weighted metrics.  
- **Tail definition**: we used an **explicit tail flag** (not recomputed). That is important because it keeps the evaluation aligned with our modeling notebook’s tail definition.

---

## A) The “main” table (all test rows, except Baseline A which cannot score all rows)

### A1) The header block (counts)
**n_train_rows = 974,651** and **n_test_rows_total = 235,624**  
- These are row-grain counts after cleaning and dropping missing keys/targets.
- **DG uses all test rows**: `n_test_rows_DG_used = 235,624`.
- **Baseline A uses only lag-available rows**: `n_test_rows_baseline_A_used = 145,076`.

✅ Immediate takeaway: **Baseline A is not a full-coverage production solution** because it only scores **61.57%** of test rows (confirmed later).

---

### A2) The “model metrics” table (the core results)
This table mixes two types of comparisons:
- **Fair comparisons** for models that score all rows (DG vs B/C/D).
- A **partly unfair comparison** for Baseline A because Baseline A is evaluated only where lag exists.

#### What each metric is telling us
- **r2_test / mae / rmse**: unweighted “overall” error.
- **log metrics (r2_test_log, mae_log, rmse_log)**: quality on log1p scale (more stable when cost is skewed).
- **weighted metrics (r2_test_w, mae_w, rmse_w)**: tail rows matter more (w_tail=10).  
  This is the closest proxy to “how bad are we in expensive cases?”
- **tail_share_of_total_sse**: fraction of total squared error contributed by tail rows.
  - Lower can mean “tail errors are under control”.
  - But it is also influenced by how well we do on non-tail rows (because total SSE is in the denominator). So treat it as supportive, not standalone.
- **tail_under_prediction_rate**: share of tail rows where pred < y.
  - High means we tend to under-benchmark high-cost rows (risk: missing anomalies).
- **tail_pred_to_y_ratio**: mean(pred)/mean(y) on tail rows.
  - Ideal is **~1.0**.
  - <1 means systematic underprediction on tail.
  - >1 means systematic overprediction on tail.
- **tail_bias**: mean(pred) − mean(y) on tail rows.
  - Ideal is **~0**.
  - Negative means underprediction, positive means overprediction.

---

### A3) Main table: what the numbers say

#### Baseline C (provider_type + POS + state mean)
- **It collapses.** r2 ~ 0.06, rmse_w ~ 563.1, tail_under_prediction_rate = 1.0, tail_pred_to_y_ratio = 0.0875.
- Interpretation: a coarse group mean is far too blunt for row-grain benchmarking.  
  It massively underpredicts high-cost rows and is unusable for anomaly surfacing.

✅ This baseline is primarily valuable as a “proof that naive group means are insufficient”.

---

#### Baseline B and Baseline D (backoff ladders)
- Both are reasonable heuristics.
- **Baseline D** (full ladder) is the stronger heuristic:
  - rmse_w = 124.60 (better than B’s 136.11)
  - tail_under_prediction_rate = 0.6829 (worse than B’s 0.5393, meaning D underpredicts tail more often)
  - tail_share_of_total_sse = 0.521 (much lower than DG’s 0.812, but see caveat below)

The key nuance:
- Baseline D is a *mean-based* backoff system, which tends to be conservative.  
  It can reduce variance in some places but often fails where relationships are non-linear or lag-driven.

---

#### DG+V3 (final)
- Strong overall: r2_test = 0.9477, rmse = 47.72.
- Tail-weighted: rmse_w = 132.05 (better than B, worse than D, not comparable to A here).
- Tail behavior:
  - tail_under_prediction_rate = 0.641 (underpredicts tail 64% of the time)
  - tail_pred_to_y_ratio = 0.959 (moderate underprediction on tail)
  - tail_bias = -68.67 (systematic underprediction in tail on average)

So why did we still prefer DG+V3 over the heuristic ladders at the decision point?
- Because this “main table” is not the full story due to Baseline A’s partial coverage and because our real architecture goal is:
  1) **use lag when available** (where it is extremely predictive),
  2) **still behave sensibly when lag is missing** (cold-start),
  3) avoid catastrophic behavior in tail (guardrails).

The “main table” is directionally useful, but it does not isolate the lag effect. That is exactly why we added the add-on.

---

## B) The “proof-style comparisons” blocks in the main cell

These blocks answer “Does DG beat each baseline on a specific metric?”

### B1) rmse_w comparisons
- DG beats B and C.
- DG loses to A and D.

Important: DG vs A is not apples-to-apples here because A uses only lag rows.

### B2) tail_share_of_total_sse comparisons
- DG beats A and C, loses to B and D.

Interpretation caution:
- A lower tail SSE share can happen if a method does relatively better on tail OR worse on non-tail.  
  We should not treat this as the single deciding metric. It is best used alongside rmse_w and tail bias/ratio.

### B3) tail_under_prediction_rate comparisons
- DG is better than C and D.
- DG is worse than A and B.

This is meaningful:
- Baseline B is better than DG at avoiding underprediction in the tail (at least on this test setting).
- Baseline A is best at tail underprediction rate, but again it only runs when lag exists.

---

## C) Lag coverage result (why Baseline A cannot be “the production model”)

**lag_coverage_pct = 61.57%**
- Only ~62% of test rows have valid lag.
- So Baseline A fundamentally cannot score ~38% of the universe.

✅ This is a core justification for the hybrid approach: we needed a system that handles both regimes.

---

## D) Apples-to-apples lag-subset table (this is the most important add-on)

This table evaluates **ALL models** on the **same subset of test rows where lag exists (mask_A)**.

### D1) What changes immediately
- DG+V3 becomes *extremely strong* on this subset:
  - r2_test = 0.9927
  - rmse = 13.91
  - rmse_w = 31.48
  - tail_pred_to_y_ratio = 0.9956 (very close to 1)
  - tail_bias = -6.31 (near zero)

That is a “lag regime” signature.  
It tells us: **DG+V3 is fully exploiting lag signal and doing so with excellent tail calibration on the lag-available rows.**

### D2) The surprising point: DG beats Baseline A on rmse_w by a lot (on lag rows)
- DG rmse_w = 31.48
- A rmse_w = 109.38

This difference is huge and usually means one of the following is true:
1) **Baseline A is predicting a different quantity** than DG’s expected_cost in this parquet, or
2) the `lag1_avg_amt` column is not aligned to this row-grain `observed_cost` target the same way we think it is (e.g., lag is at a different aggregation), or
3) Baseline A is valid but has a systematic tail overprediction on this evaluation definition.

Our deltas table reinforces (3):
- Baseline A tail_pred_to_y_ratio = 1.078 (overpredicts tail by ~7.8%)
- Baseline A tail_bias = +111.76 (strong overprediction)
- DG tail_pred_to_y_ratio ~ 0.996, tail_bias ~ -6.3 (near neutral)

Interpretation:
- Baseline A is powerful, but here it looks like it **overshoots** high-cost rows on the lag subset.
- DG+V3 is more calibrated on the tail within lag rows.

### D3) Why B matches A exactly on the lag subset in our output
In the lag subset table:
- Baseline B metrics are identical to Baseline A.

That is expected given our Baseline B definition:
- Baseline B uses lag first, then backoff.
- On rows where lag exists (mask_A), B effectively reduces to A.

✅ This is a nice validation that our comparison logic is working as intended.

---

## E) How to summarize the decision logic (the “why DG+V3” argument)

### The conclusion our output supports (clean version)
1) **Lag is a dominant predictor when available**, and any strong production system must exploit it.
2) **Lag is missing for ~38% of rows**, so a lag-only baseline cannot serve the full universe.
3) On the lag-available subset, **DG+V3 is extremely accurate and well-calibrated on tail rows**, and it actually outperforms the lag-only baseline here on tail-weighted error and tail bias.
4) On the full universe, DG+V3 is competitive with heuristic backoff approaches while providing:
   - a single consistent expected-cost score for every row,
   - guardrail-bounded behavior,
   - and compatibility with downstream anomaly surfacing that relies on consistent residual/log(OE) behavior.

### The honest nuance we should keep
- Baseline D looks better than DG on **rmse_w** in the full-universe table (124.6 vs 132.0).  
  That tells us the backoff ladder is a strong heuristic baseline.
- However, the lag-subset result is the stronger evidence for our “hybrid architecture” claim:
  DG is excellent when lag exists AND it still runs when lag does not.

If we phrase it like this, it is both rigorous and defensible:
- “Lag-only baselines can look great where lag exists, but they do not cover the whole problem. DG+V3 combines lag exploitation with robust cold-start behavior and guardrails, producing a single expected-cost benchmark for every row while staying tail-aware.”

---

## F) One important sanity note to consider (not required, but worth knowing)
Given how dramatically DG beats Baseline A on the lag subset, it is worth confirming that:
- `lag1_avg_amt` is truly the correct lag feature aligned to `observed_cost` at the row grain (NPI×HCPCS×POS×Year).
- In some pipelines, lag might be defined at a slightly different aggregation, or may represent a different “amount” unit than observed_cost in eval_scored.

We don’t need to change anything now. Just be aware: that massive gap suggests DG is using more than lag (which it is), and also that Baseline A’s lag value may be systematically too high in the tail under this evaluation definition.
